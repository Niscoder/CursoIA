## Limpeza e transformação dos dados

Site oficial do [pandas](https://pandas.pydata.org/).

Importando o pandas:

```python
import pandas as pd
```

Carregando o arquivo `csv` como `DataFrame`:

```python
# Função para carregar arquivo
data = pd.read_csv('./dados.csv')

# Printando informações gerais da base de dados
print(data.info())
```

Simplificando os nomes das colunas:

```python
# Removendo a coluna que temos certeza que não será utilizada
data = data.drop(columns=["Carimbo de data/hora"])

# Print ordenado dos nomes das colunas
print(data.columns)
```

```python
# Vetor com os nomes desejados das colunas
columns_name = ["idade", "universidade", "curso", "periodo",
                "turno", "ling_prog", "area_atuacao", "art_banda",
                "filme_serie", "comida_fav", "hobby", "cidade_nat",
                "atv_fisica", "altura"]

# Criando dicionário com chave nome atual e valor novo nome para a renomeação
columns_rename = {}
for idx, column in enumerate(data.columns):
    columns_rename[column] = columns_name[idx]

print(columns_rename)

# Renomenado as colunas
data = data.rename(columns=columns_rename)
print(data.info())
```

Análise, limpeza e conversão para inteiro dos valores de `idade`:

```python
# Printando a quantidade de cada valor existente na coluna idade
print(data["idade"].value_counts())
```

```python
# Substituição direta do valor incompatível
data["idade"] = data["idade"].replace({"19 anos": "19"})

# Conversão direta de todos os valores da coluna para inteiro
data["idade"] = data["idade"].astype(int)

print(data["idade"].value_counts())
```

```python
print(data.info())
```

Análise, limpeza e conversão para inteiro dos valores de `altura`:

```python
# Printando a quantidade de cada valor existente na coluna altura
print(data["altura"].value_counts())
```

```python
# Função personalizada para limpeza e conversão dos dados
def transform_idade(value: str):
    # Verifica se o valor tem somente número
    if not value.isdigit():
        value = value.lower()
        value = value.replace(".", "").replace(",", "").replace("m", "").replace("c", "")
        value = value.strip()
        if len(value) == 2:
            value = f"{value}0"

        value = value[:3]
    # Conversão de valor direta no retorno
    return int(value)

# Transformação dos valores através da função personalizada
data["altura"] = data["altura"].apply(transform_idade)

print(data["altura"].value_counts())
```

```python
print(data.info())
```

Análise e conversão para booleano dos valores de `atv_fisica`:

```python
# Printando a quantidade de cada valor existente na coluna atv_fisica
print(data["atv_fisica"].value_counts())
```

```python
# Substituição direta do valor incompatível
data["atv_fisica"] = data["atv_fisica"].replace({"Sim": True, "Não": False})

print(data["atv_fisica"].value_counts())
```

```python
print(data.info())
```

Análise, limpeza e conversão para inteiro dos valores de `periodo`:

```python
# Printando a quantidade de cada valor existente na coluna periodo
print(data["periodo"].value_counts())
```

```python
# Função personalizada para limpeza e conversão dos dados
def transform_periodo(value: str):
    value = value.lower()
    value = value.replace("semestre", "").replace("bimestre", "")
    value = value.strip()
    value = value.replace("segundo", "2").replace("sexto", "6")
    # Conversão de valor direta no retorno
    return int(value[0])

# Transformação dos valores através da função personalizada
data["periodo"] = data["periodo"].apply(transform_periodo)

print(data["periodo"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `curso`:

```python
# Printando a quantidade de cada valor existente na coluna curso
print(data["curso"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_curso(value: str):
    value = value.lower()
    if "dado" in value:
        value = "ciencia de dados e IA"

    elif "software" in value:
        value = "engenharia de software"

    elif "informa" in value or "s.i" in value:
        value = "sistemas de informacao"

    elif "computação" in value:
        value = "ciencia da computacao"

    elif "desenv" in value or "ads" in value:
        value = "ads"

    return value

# Transformação dos valores através da função personalizada
data["curso"] = data["curso"].apply(transform_curso)

print(data["curso"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `universidade`:

```python
# Printando a quantidade de cada valor existente na coluna universidade
print(data["universidade"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_universidade(value: str):
    value = value.lower().strip()
    if "unifil" in value:
        value = "unifil"

    elif "utfpr" in value:
        value = "utfpr"

    return value

# Transformação dos valores através da função personalizada
data["universidade"] = data["universidade"].apply(transform_universidade)

print(data["universidade"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `ling_prog`:

```python
# Printando a quantidade de cada valor existente na coluna ling_prog
print(data["ling_prog"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_ling_prog(value: str):
    value = value.strip().lower()
    value = value.split(",")[0]
    value = value.split()[0]
    if "node" in value:
        value = "javascript"

    return value

# Transformação dos valores através da função personalizada
data["ling_prog"] = data["ling_prog"].apply(transform_ling_prog)

print(data["ling_prog"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `comida_fav`:

```python
# Printando a quantidade de cada valor existente na coluna comida_fav
print(data["comida_fav"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_comida_fav(value: str):
    value = value.strip().lower()
    if "hamb" in value or "lanche" in value:
        value = "lanche"

    elif "jap" in value:
        value = "japonesa"

    elif "churr" in value:
        value = value

    elif not ("." in value or "strog" in value or "comid" in value or "fric" in value):
        value = "massa"

    else:
        value = "outros"

    return value

# Transformação dos valores através da função personalizada
data["comida_fav"] = data["comida_fav"].apply(transform_comida_fav)

print(data["comida_fav"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `hobby`:

```python
# Printando a quantidade de cada valor existente na coluna hobby
print(data["hobby"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_hobby(value: str):
    value = value.lower()
    if "," in value:
        value = value.split(",")[0]

    sports = ["fute", "bola", "vôlei", "espor", "pipa", "patins"]
    for sport in sports:
        if sport in value:
            return "esporte"

    arts = ["jog", "valora", "fifa", "rpg", "tcg", "game", "level"]
    for art in arts:
            if art in value:
                return "jogos"

    arts = ["ler", "desen", "ediç", "vocal", "bateria", "artes", "aula"]
    for art in arts:
            if art in value:
                return "artes"

    return "outros"

# Transformação dos valores através da função personalizada
data["hobby"] = data["hobby"].apply(transform_hobby)

print(data["hobby"].value_counts())
```

```python
print(data.info())
```

Análise e transformação dos valores de `area_atuacao`:

```python
# Printando a quantidade de cada valor existente na coluna area_atuacao
print(data["area_atuacao"].value_counts())
```

```python
# Função personalizada para transformação dos dados
def transform_area_atuacao(value: str):
    value = value.lower()
    ia = ["a.i", "cien", "dado", "data", "artific", "machi", "gener", "ia"]
    for area in ia:
        if area in value:
            return "ciencia_dados/ia"

    dev = ["desenv", "dev", "front", "back"]
    for area in dev:
        if area in value:
            return "dev"

    manag = ["gest", "product", "agile", "ceo", "proje"]
    for area in manag:
        if area in value:
            return "gestao/diretoria"

    return "n/d"

# Transformação dos valores através da função personalizada
data["area_atuacao"] = data["area_atuacao"].apply(transform_area_atuacao)

print(data["area_atuacao"].value_counts())
```

```python
print(data.info())
```
