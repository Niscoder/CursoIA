```python
import pandas as pd

# Carregando dados csv como dataframe
data = pd.read_csv('./dados.csv')

#printando as infos das colunas
print(data.info())
```

```python
# Remover coluna desnecessária da base de dados
data = data.drop(columns=["Carimbo de data/hora"])

print(data.info())
```

```python
# Seleciona a coluna Idade e realiza a contagem dos valores únicos
vu = data["Idade"].value_counts()
print(vu)
```

```python
# Transformando os dados da coluna idade, substituição de valores
data["Idade"] = data["Idade"].replace({"19 anos": "19"})
vu = data["Idade"].value_counts()
print(vu)
```

```python
# Convertendo os valores da coluna Idade para inteiro
data["Idade"] = data["Idade"].astype(int)

print(data.info())
```

```python
#Verificando os valores da coluna Altura
print(data["Altura"].value_counts())
```

```python
# Função personalizada de transformação dos dados da coluna Altura
def transformacao_altura(texto: str):
    if not texto.isdigit():
        texto = texto.lower()
        texto = texto.replace(".", "").replace(",", "").replace("m", "").replace("c", "")
        texto = texto.strip()
        if len(texto) == 2:
            texto = f"{texto}0"

        texto = texto[:3]

    return int(texto)

#Aplicando a função de transformação da Altura
data["Altura"] = data["Altura"].apply(transformacao_altura)
vt = data["Altura"].value_counts()
print(vt)

```

```python
# Seleciona a coluna Semestre/Período e realiza a contagem dos valores únicos
print(data["Semestre/Período"].value_counts())
```

```python
# Função personalizada de transformação dos dados da coluna Semestre/Período
def transformacao_semestre(texto: str):
    texto = texto.lower().replace("semestre", "").replace("bimestre", "").strip()
    texto = texto.replace("segundo", "2").replace("sexto", "6")
    return int(texto[0])

vt = data["Semestre/Período"].apply(transformacao_semestre)
print(vt.value_counts())
```

```python
data = data.rename(columns={"Semestre/Período": "periodo"})
data["periodo"] = vt
print(data.info())
```

```python
def transformacao_instituicao(text: str):
    text = text.lower().strip()
    if "unifil" in text:
        text = "unifil"

    elif "utfpr" in text:
        text = "utfpr"

    return text

vt = data["Instituição de ensino"].apply(transformacao_instituicao)
print(vt.value_counts())
```

```python
column_name = "Comida favorita (Hamburger, pizza, japonesa, alemã)"

print(data[column_name].value_counts())
```

```python
def transformacao_comfav(text: str):
    text = text.strip().lower()

    if "hamb" in text or "lanche" in text:
        text = "lanche"

    elif "jap" in text:
        text = "japonesa"

    elif "churr" in text:
        text = text

    elif not ("." in text or "strog" in text or "comid" in text or "fric" in text):
        text = "massa"

    else:
        text = "outros"

    return text

vt = data[column_name].apply(transformacao_comfav)
print(vt.value_counts())
```

```python
print(data["Hobby"].value_counts())
```

```python
def transformacao_hobby(text: str):
    text = text.lower()
    if "," in text:
        text = text.split(",")[0]

    esportes = ["fute", "bola", "vôlei", "espor", "pipa", "patins"]
    for esporte in esportes:
        if esporte in text:
            text = "esporte"
            break

    jogos = ["jog", "valora", "fifa", "rpg", "tcg", "game", "level"]
    for jogo in jogos:
            if jogo in text:
                text = "jogos"
                break

    jogos = ["ler", "desen", "ediç", "vocal", "bateria", "artes", "aula"]
    for jogo in jogos:
            if jogo in text:
                text = "artes"
                break

    if not text in ["jogos", "esporte", "artes"]:
        text = "outros"

    return text

vt = data["Hobby"].apply(transformacao_hobby)
print(vt.value_counts())
```

```python
data["Hobby"] = vt

print(data.info())
```

```python
column_name = "Área de atuação desejada"
data = data.rename(columns={column_name: "area_atuacao"})

print(data["area_atuacao"].value_counts())
```

```python
def transformacao_area(text: str):
    text = text.lower()


    areas_ia = ["a.i", "cien", "dado", "data", "artific", "machi", "gener", "ia"]
    for area in areas_ia:
            if area in text:
                text = "ciencia_dados/ia"
                break

    areas_dev = ["desenv", "dev", "front", "back"]
    for area in areas_dev:
            if area in text:
                text = "dev"
                break

    areas_gest = ["gest", "product", "agile", "ceo", "proje"]
    for area in areas_gest:
            if area in text:
                text = "gestao/diretoria"
                break

    if not text in ["ciencia_dados/ia", "gestao/diretoria", "dev"]:
        text = "n/d"

    return text

vt = data["area_atuacao"].apply(transformacao_area)
print(vt.value_counts())
```

```python
data["area_atuacao"] = vt

print(data.info())
```
