```python
def myfunc(a):
    if a:
        return 10

print(myfunc(False))
```

```python
def verificar_idade():
    idade = 20

    if idade >= 18:
        return

    return "Menor de idade"

print(verificar_idade())
```

```python
def saudacao(nome):
    print(f"Olá, {nome}!")

saudacao("João")
saudacao(2)
```

```python
def soma(a: int, b: int):
    if type(a) == int and type(b) == int:
        return a + b

    return 0

print(soma("João", 3))
```

```python
def apresentar(nome, idade):
    print(f"Nome: {nome}")
    print(f"Idade: {idade}")

apresentar(25, "João")
```

```python
def apresentar(nome, idade):
    print(f"Nome: {nome}")
    print(f"Idade: {idade}")

apresentar(nome="João", idade=25)
apresentar(nome="Pedro", idade=30)
```

```python
def apresentar(nome, idade, carteira):
    print(f"Nome: {nome}")
    print(f"Idade: {idade}")
    print(f"Tem carteira?: {carteira}")

apresentar("Pedro", carteira=True, idade=25)
apresentar("Pedro", 25, carteira=False)
```

```python
def operacao(num_a, num_b, div=1):
    return (num_a + num_b) / div

print(operacao(4, 2))
print(operacao(4, 2, 3))
print(operacao(div=2, num_b=10, num_a=4))
```

```python
def operacao(num_b, num_a=2, div=1):
    return (num_a + num_b) / div
```

```python
def soma(*args):
    print(args[1])

soma(1)
soma(1, 2)
soma(1, 4, 5)
soma()
```

```python
def apresentar(**kwargs):
    print(kwargs["nome"])

apresentar(nome="João", idade=25, cidade="Londrina")
```

```python
def exemplo(*args, **kwargs):
    print("args:", args)
    print("**kwargs:", kwargs)

exemplo(10, 20, 30, nome="João", idade=25)
```

```python
def exemplo(nome, *args, idade=18, **kwargs):
    print("nome:", nome)
    print("idade:", idade)
    print("args:", args)
    print("**kwargs:", kwargs)

exemplo(
    "João",
    "Python",
    "Machine Learning",
    idade=25,
    cidade="Londrina",
    curso="TI"
)
```
