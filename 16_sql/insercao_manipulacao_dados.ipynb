# INSERT

O comando `INSERT` adiciona registros a uma tabela.

A forma básica é:

```sql
INSERT INTO tabela (coluna1, coluna2, coluna3)
VALUES (valor1, valor2, valor3);
```

Por exemplo:

```sql
INSERT INTO clientes (nome, email)
VALUES ('Ana Silva', 'ana@email.com');
```

Como o `id` foi definido como `IDENTITY`, não precisamos informar seu valor.

O banco irá gerar o identificador automaticamente.


## Inserindo vários registros

Podemos inserir vários registros em um único comando:

```sql
INSERT INTO clientes (nome, email)
VALUES
    ('Carlos Souza', 'carlos@email.com'),
    ('Mariana Oliveira', 'mariana@email.com'),
    ('João Santos', 'joao@email.com'),
    ('Fernanda Lima', 'fernanda@email.com');
```

Essa abordagem é normalmente mais prática do que executar um `INSERT` separado para cada registro.


## Inserindo utilizando valores DEFAULT

Quando uma coluna possui `DEFAULT`, podemos simplesmente não informar essa coluna.

Considere:

```sql
CREATE TABLE produtos (
    id INTEGER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    nome VARCHAR(100) NOT NULL,
    preco NUMERIC(10,2) NOT NULL CHECK (preco >= 0),
    estoque INTEGER NOT NULL DEFAULT 0 CHECK (estoque >= 0)
);
```

Podemos inserir:

```sql
INSERT INTO produtos (nome, preco)
VALUES ('Teclado', 150.00);
```

Nesse caso, `estoque` receberá automaticamente `0`.


# SELECT

O `SELECT` é utilizado para consultar dados.

Para consultar todas as colunas:

```sql
SELECT *
FROM clientes;
```

O `*` representa todas as colunas.

Embora seja bastante útil durante o aprendizado e exploração dos dados, em consultas de aplicações é frequentemente melhor especificar apenas as colunas necessárias.


## Selecionando colunas específicas

Podemos escolher quais colunas queremos visualizar:

```sql
SELECT nome, email
FROM clientes;
```

Também podemos utilizar aliases:

```sql
SELECT
    nome AS nome_cliente,
    email AS email_cliente
FROM clientes;
```

O `AS` permite definir um nome para a coluna retornada pela consulta.


# WHERE

O `WHERE` permite filtrar os registros retornados.

Por exemplo:

```sql
SELECT *
FROM clientes
WHERE id = 2;
```

Podemos utilizar operadores de comparação:

```text
=     igual
<>    diferente
>     maior
<     menor
>=    maior ou igual
<=    menor ou igual
```

Exemplo:

```sql
SELECT *
FROM produtos
WHERE preco > 100;
```


## AND e OR

Podemos combinar condições utilizando `AND` e `OR`.

### AND

Todas as condições precisam ser verdadeiras:

```sql
SELECT *
FROM produtos
WHERE preco >= 100
  AND estoque > 0;
```

### OR

Pelo menos uma das condições precisa ser verdadeira:

```sql
SELECT *
FROM produtos
WHERE preco < 50
   OR estoque = 0;
```


## NOT

Também podemos negar uma condição:

```sql
SELECT *
FROM produtos
WHERE NOT estoque = 0;
```

Uma forma equivalente e normalmente mais clara seria:

```sql
SELECT *
FROM produtos
WHERE estoque <> 0;
```


# IN

O operador `IN` permite verificar se um valor pertence a uma lista.

Em vez de:

```sql
SELECT *
FROM clientes
WHERE id = 1
   OR id = 3
   OR id = 5;
```

podemos escrever:

```sql
SELECT *
FROM clientes
WHERE id IN (1, 3, 5);
```

Isso é especialmente útil quando temos vários valores possíveis.


# BETWEEN

`BETWEEN` permite verificar se um valor está dentro de um intervalo.

Por exemplo:

```sql
SELECT *
FROM produtos
WHERE preco BETWEEN 50 AND 200;
```

O intervalo inclui os valores das extremidades.

Nesse exemplo, produtos com preço exatamente `50` ou `200` também serão considerados.


# LIKE

O operador `LIKE` permite realizar comparações utilizando padrões de texto.

O caractere `%` representa qualquer sequência de caracteres.

Por exemplo:

```sql
SELECT *
FROM clientes
WHERE nome LIKE 'Ana%';
```

Essa consulta encontra nomes que começam com `Ana`.

Outro exemplo:

```sql
SELECT *
FROM clientes
WHERE email LIKE '%@gmail.com';
```

Aqui procuramos e-mails que terminam com `@gmail.com`.

O caractere `_` representa exatamente um caractere:

```sql
SELECT *
FROM clientes
WHERE nome LIKE 'Jo_o';
```


# 8. NULL

`NULL` representa a ausência de um valor.

É importante entender que:

```text
NULL
```

não é a mesma coisa que:

```text
''
```

Uma string vazia é um texto com zero caracteres. `NULL` significa que não existe um valor.

Para verificar `NULL`, não devemos utilizar:

```sql
WHERE coluna = NULL
```

Devemos utilizar:

```sql
WHERE coluna IS NULL;
```

E para verificar valores que não são `NULL`:

```sql
WHERE coluna IS NOT NULL;
```


# ORDER BY

`ORDER BY` permite ordenar os resultados.

Por exemplo:

```sql
SELECT *
FROM produtos
ORDER BY preco;
```

Por padrão, a ordenação é crescente (`ASC`).

Podemos escrever explicitamente:

```sql
SELECT *
FROM produtos
ORDER BY preco ASC;
```

Para ordenar de forma decrescente:

```sql
SELECT *
FROM produtos
ORDER BY preco DESC;
```


## Ordenando por várias colunas

Podemos utilizar mais de uma coluna:

```sql
SELECT *
FROM produtos
ORDER BY
    preco DESC,
    nome ASC;
```

Nesse caso, o preço é utilizado primeiro.

Quando dois produtos possuem o mesmo preço, o nome é utilizado para desempatar.


# LIMIT e OFFSET

`LIMIT` limita a quantidade de registros retornados.

```sql
SELECT *
FROM produtos
LIMIT 10;
```

Podemos combinar com `ORDER BY`:

```sql
SELECT *
FROM produtos
ORDER BY preco DESC
LIMIT 5;
```

Essa consulta retorna os cinco produtos de maior preço.

`OFFSET` permite ignorar uma quantidade de registros:

```sql
SELECT *
FROM produtos
ORDER BY id
LIMIT 10
OFFSET 10;
```

Esse mecanismo é bastante utilizado para implementar paginação.


# UPDATE

O `UPDATE` altera registros existentes.

Estrutura:

```sql
UPDATE tabela
SET coluna = valor
WHERE condição;
```

Por exemplo:

```sql
UPDATE produtos
SET preco = 180.00
WHERE id = 1;
```

O produto de `id = 1` terá seu preço alterado.


## Alterando várias colunas

Podemos alterar várias colunas no mesmo comando:

```sql
UPDATE produtos
SET
    preco = 180.00,
    estoque = 25
WHERE id = 1;
```


## O cuidado mais importante com UPDATE

Considere:

```sql
UPDATE produtos
SET preco = 180.00;
```

Sem `WHERE`, **todos os registros** da tabela serão atualizados.

Isso pode ser intencional, mas frequentemente é um erro.

Uma boa prática durante o aprendizado é executar primeiro o `SELECT` correspondente:

```sql
SELECT *
FROM produtos
WHERE id = 1;
```

Depois executar o `UPDATE`:

```sql
UPDATE produtos
SET preco = 180.00
WHERE id = 1;
```

E então verificar novamente:

```sql
SELECT *
FROM produtos
WHERE id = 1;
```


# DELETE

O comando `DELETE` remove registros.

Estrutura:

```sql
DELETE FROM tabela
WHERE condição;
```

Por exemplo:

```sql
DELETE FROM produtos
WHERE id = 5;
```

O produto de `id = 5` será removido.


## O cuidado mais importante com DELETE

Assim como no `UPDATE`, devemos ter muito cuidado com a ausência de `WHERE`.

```sql
DELETE FROM produtos;
```

Esse comando remove **todos os registros** da tabela.

A tabela continua existindo, mas seus registros serão removidos.

Uma prática segura é verificar primeiro:

```sql
SELECT *
FROM produtos
WHERE id = 5;
```

Depois:

```sql
DELETE FROM produtos
WHERE id = 5;
```
