## Operadores lógicos

Operadores lógicos são utilizados para combinar ou inverter expressões que possuem resultado booleano (`True` ou `False`).

### Operador `and`

Retorna `True` quando todas as condições são verdadeiras.

```python
idade = 25
possui_carteira = True

resultado = idade >= 18 and possui_carteira

print(resultado)
```

### Operador `or`

Retorna `True` quando pelo menos uma das condições é verdadeira.

```python
idade = 16
acompanhado = True

resultado = idade >= 18 or acompanhado

print(resultado)
```

### Operador `not`

Inverte o valor lógico de uma expressão.

```python
possui_carteira = False

resultado = not possui_carteira

print(resultado)
```

## Operadores de comparação

Operadores de comparação permitem comparar valores e sempre retornam um `bool`.

| Operador | Significado    |
| -------- | -------------- |
| `==`     | Igual          |
| `!=`     | Diferente      |
| `>`      | Maior que      |
| `<`      | Menor que      |
| `>=`     | Maior ou igual |
| `<=`     | Menor ou igual |

```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

## Estruturas condicionais

Estruturas condicionais permitem executar diferentes trechos de código de acordo com uma condição.

### `if`

O `if` executa um bloco de código somente quando sua condição é verdadeira.

```python
idade = 20

if idade >= 18:
    print("Maior de idade")
```

A indentação define quais instruções pertencem ao bloco condicional.

### `else`

O `else` define o bloco que será executado quando a condição do `if` for falsa.

```python
idade = 16

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
```

### `elif`

O `elif` permite verificar uma nova condição quando as condições anteriores forem falsas.

```python
idade = 16

if idade >= 18:
    print("Maior de idade")
elif idade >= 12:
    print("Adolescente")
else:
    print("Criança")
```

É possível utilizar vários `elif` em uma mesma estrutura.

## Combinando condições

Operadores lógicos podem ser utilizados para criar condições mais complexas.

```python
idade = 25
possui_carteira = True

if idade >= 18 and possui_carteira:
    print("Pode dirigir")
else:
    print("Não pode dirigir")
```

Também é possível utilizar parênteses para definir a ordem das operações lógicas:

```python
idade = 20
estudante = True
trabalha = False

if idade >= 18 and (estudante or trabalha):
    print("Condição atendida")
```

## Valores considerados verdadeiros ou falsos

Em Python, alguns valores são avaliados como `False` em uma condição, como `0`, `None` e strings vazias. Outros valores são avaliados como `True`.

```python
valor = 10

if valor:
    print("O valor é considerado verdadeiro")
```

Uma forma explícita de verificar essa conversão é utilizar `bool()`:

```python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
```
