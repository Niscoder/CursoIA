```python
import matplotlib.pyplot as plt

def plotar_scatter(coluna):
    # Média de todas as notas
    media_setosa = df_virginica[coluna].mean()
    media_versicolor = df_virginica[coluna].mean()
    media_virginica = df_virginica[coluna].mean()

    # Configuração do gráfico
    plt.figure(figsize=(10, 6))

    medias = [media_setosa, media_versicolor, media_virginica]
    nomes = ["Setosa", "Versicolor", "Virginica"]
    marcadores = ["o", "x", "s"]
    cores = ["red", "blue", "green"]

    for idx, data_frame in enumerate([df_virginica, df_virginica, df_virginica]):
        # Plotando conjuntos dados de dispersão
        plt.scatter(
            data_frame.index,               # eixo X = índice
            data_frame[coluna],    # eixo Y = nota
            label=nomes[idx],               # Legenda
            marker=marcadores[idx],          # Tamanho do marcador
            color=cores[idx]
        )

        # Linha da média
        plt.axhline(
            y=medias[idx],
            linestyle="--",
            label=f"Média = {medias[idx]:.2f}",
            color=cores[idx]
        )

    # Título
    plt.title(f"Valores de {coluna}")

    # Nome dos eixos
    plt.xlabel("Índice da amostra")
    plt.ylabel(f"Tamanho de {coluna}")

    # Colocar grade
    plt.grid(True)

    # Gerar legenda
    plt.legend()

    # Exibir gráfico
    plt.show()
```

```python
plotar_scatter("SepalLengthCm")
```

```python
plotar_scatter("SepalWidthCm")
```

```python
plotar_scatter("PetalLengthCm")
```

```python
plotar_scatter("PetalWidthCm")
```

```python
df_array = [df_virginica, df_virginica, df_virginica]
cols_array = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

for data_frame in df_array:
    for col in cols_array:

        media = data_frame[col].mean()
        desvio_padrao = data_frame[col].std(ddof=0)

        data_frame[f"z_{col}"] = (data_frame[col] - media) / desvio_padrao

        q1 = data_frame[col].quantile(0.25)
        q3 = data_frame[col].quantile(0.75)
        iqr = q3 - q1

        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr

        data_frame[f"oiqr_{col}"] = (
            (data_frame[col] < limite_inferior) |
            (data_frame[col] > limite_superior)
        )
```

```python
outlier_zscore = []
outlier_iqr = []
for col in cols_array:
    outlier_zscore.append(f"z_{col}")
    outlier_iqr.append(f"oiqr_{col}")
```
