## Boas práticas

Ao escrever código, não basta que ele funcione. Também é importante que seja fácil de ler, entender e manter.

Python possui algumas convenções de escrita que ajudam a manter o código consistente. Essas convenções são principalmente recomendações e não regras obrigatórias da linguagem.

Uma das principais é a convenção de nomes.

### Variáveis e funções — `snake_case`

Para variáveis e funções, a convenção utilizada em Python é o `snake_case`.

Nesse padrão:

* utiliza-se apenas letras minúsculas;
* palavras são separadas por `_`.

Exemplos:

```python
nome_usuario = "Gustavo"
idade_usuario = 25
quantidade_produtos = 10
```

Para funções:

```python
def calcular_media():
    ...
```

```python
def obter_nome_usuario():
    ...
```

Evite utilizar espaços ou misturar diferentes padrões:

```python
# Evitar
nomeUsuario = "Gustavo"
NomeUsuario = "Gustavo"
nome_usuario = "Gustavo"  # snake_case
```

O `snake_case` também deve ser utilizado para os métodos das classes:

```python
class Pessoa:
    def obter_nome(self):
        return self.nome
```

### Classes — `CamelCase`

Para classes, a convenção é utilizar `CamelCase`, também chamada de PascalCase.

Nesse padrão:

* cada palavra começa com letra maiúscula;
* não são utilizados `_` entre as palavras.

Exemplos:

```python
class Pessoa:
    ...


class ContaBancaria:
    ...


class CadastroUsuario:
    ...
```

Observe a diferença:

```text
Variável:  nome_usuario
Função:    obter_nome_usuario()
Classe:    CadastroUsuario
```

### Constantes — letras maiúsculas

Valores que representam constantes devem ser escritos utilizando letras maiúsculas, com palavras separadas por `_`.

```python
PI = 3.14159
TAXA_JUROS = 0.05
LIMITE_TENTATIVAS = 3
```

Isso indica que o valor deve ser tratado como uma constante e não alterado durante a execução do programa.

É importante entender que isso também é uma convenção. Python não impede que o valor seja alterado:

```python
LIMITE_TENTATIVAS = 3

LIMITE_TENTATIVAS = 5
```

O código funciona, mas a utilização de nomes em maiúsculas comunica aos outros desenvolvedores que aquele valor deve ser tratado como constante.

### Métodos e atributos privados — `_`

Quando um método ou atributo começa com `_`, isso indica que ele é considerado interno ou privado por convenção.

Por exemplo:

```python
class Pessoa:
    def __init__(self, nome):
        self.nome = nome
        self._documento = None

    def _validar_documento(self):
        ...
```

Nesse exemplo:

```text
nome                 → atributo público
_documento           → atributo interno
_validar_documento() → método interno
```

O `_` não impede o acesso ao atributo ou método:

```python
pessoa = Pessoa("Gustavo")

print(pessoa._documento)
pessoa._validar_documento()
```

O objetivo é comunicar a intenção de uso, aquele elemento faz parte da implementação interna da classe e não deveria ser utilizado diretamente por código externo.

### Evitando nomes pouco descritivos

Os nomes devem deixar claro o que representam.

Evite nomes genéricos quando um nome mais descritivo for possível:

```python id="w5n8qa"
# Evitar
x = 10
y = 20
z = x + y
```

Prefira:

```python
preco = 10
quantidade = 20
total = preco + quantidade
```

Da mesma forma, funções devem indicar a ação que executam:

```python
# Evitar
def dados():
    ...


# Preferir
def obter_dados_usuario():
    ...
```

Bons nomes reduzem a necessidade de comentários para explicar o que o código está fazendo.

### Resumo das convenções

As principais convenções de nomes apresentadas são:

```text
Variáveis       → snake_case
Funções         → snake_case
Métodos         → snake_case
Classes         → CamelCase
Constantes      → MAIÚSCULAS_COM_UNDERSCORE
Privados        → _nome
```

Exemplo reunindo as convenções:

````python
TAXA_DESCONTO = 0.10


class CalculadoraPreco:

    def __init__(self, preco):
        self.preco = preco
        self._desconto = TAXA_DESCONTO

    def calcular_preco_final(self):
        return self.preco * (1 - self._desconto)
````

Essas convenções tornam o código mais consistente, legível e previsível, especialmente quando várias pessoas trabalham no mesmo projeto.
