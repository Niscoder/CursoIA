## Listas

Uma lista é uma estrutura utilizada para armazenar múltiplos valores em uma única variável.

Os elementos de uma lista são ordenados e podem ser acessados individualmente por meio de um índice.

```python
nomes = ["Gabriel", "Maria", "João", "Carlos"]
```

Os índices começam em `0`:

```python
print(nomes[0])
print(nomes[1])
print(nomes[2])
```

### Listas podem conter qualquer objeto

Uma lista pode conter objetos de diferentes tipos. Inclusive, uma lista pode conter outras listas.

```python
dados = ["Gabriel", 25, 1.75, True, ["Python", "Machine Learning"]]

print(dados)
```

Nesse exemplo, a lista contém objetos `str`, `int`, `float`, `bool` e outra `list`.

Uma lista também pode conter outras listas em diferentes níveis:

```python
matriz = [[1, 2, 3], [4, 5, 6],[7, 8, 9]]

print(matriz[0])
print(matriz[0][1])
```

O primeiro índice acessa a lista interna e o segundo acessa o elemento dentro dela.

### Alterando elementos

Listas são mutáveis, portanto seus elementos podem ser alterados após sua criação.

```python
nomes = ["Gabriel", "Maria", "João"]

nomes[1] = "Ana"

print(nomes)
```

### Slicing

O *slicing* permite obter uma parte de uma lista utilizando a sintaxe `lista[inicio:fim:passo]`. O índice `inicio` é incluído, enquanto `fim` não é incluído.

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[1:4])
```

O resultado contém os elementos dos índices `1` até `3`.

O início ou o fim podem ser omitidos:

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[:3])
print(numeros[3:])
print(numeros[:])
```

Também podemos definir um `passo`:

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[0:6:2])
```

Nesse caso, serão selecionados os elementos pulando de dois em dois.

### Índices e steps negativos

Índices negativos permitem acessar elementos a partir do final da lista:

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[-1])
print(numeros[-2])
```

Também é possível utilizar índices negativos no início ou no fim do *slice*:

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[1:-1])
```

Nesse caso, o *slice* começa no índice `1` e termina antes do último elemento.

O `step` também pode ser negativo. Isso permite percorrer a lista no sentido inverso:

```python
numeros = [0, 1, 2, 3, 4, 5]

print(numeros[::-1])
print(numeros[4:1:-1])
```

### Adicionando elementos

O método `append()` adiciona um elemento ao final da lista.

```python
nomes = ["Gabriel", "Maria"]

nomes.append("João")

print(nomes)
```

O método `insert()` permite adicionar um elemento em uma posição específica.

```python
nomes = ["Gabriel", "Maria"]

nomes.insert(1, "João")

print(nomes)
```

### Adicionando múltiplos elementos

O método `extend()` adiciona os elementos de outra sequência ao final da lista.

```python
nomes = ["Gabriel", "Maria"]

nomes.extend(["João", "Carlos"])

print(nomes)
```

Diferente do `append()`, que adiciona um único objeto como um elemento da lista, o `extend()` adiciona cada elemento da sequência individualmente.

```python
nomes = ["Gabriel", "Maria"]

nomes.append(["João", "Carlos"])

print(nomes)

nomes = ["Gabriel", "Maria"]

nomes.extend(["João", "Carlos"])

print(nomes)
```

### Removendo elementos

O método `remove()` remove um elemento pelo seu valor.

```python
nomes = ["Gabriel", "Maria", "João"]

nomes.remove("Maria")

print(nomes)
```

O método `pop()` remove um elemento pelo seu índice e retorna o elemento removido, ou o último elemento caso não seja informado o índice.

```python
nomes = ["Gabriel", "Maria", "João"]

nome = nomes.pop(1)

print(nome)
print(nomes)

nome = nomes.pop()

print(nome)
print(nomes)
```

### Contando elementos

O método `count()` retorna a quantidade de vezes que determinado valor aparece na lista.

```python
numeros = [1, 2, 2, 3, 2, 4]

quantidade = numeros.count(2)

print(quantidade)
```

### Tamanho da lista

A função `len()` retorna a quantidade de elementos de uma lista.

```python
nomes = ["Gabriel", "Maria", "João"]

print(len(nomes))
```

### Copiando uma lista

Atribuir uma lista para outra variável não cria uma nova lista. As duas variáveis passam a referenciar o mesmo objeto.

```python
nomes = ["Gabriel", "Maria", "João"]

outros_nomes = nomes

outros_nomes[0] = "Carlos"

print(nomes)
print(outros_nomes)
```

Para criar uma nova lista com os mesmos elementos, podemos utilizar o método `copy()`:

```python
nomes = ["Gabriel", "Maria", "João"]

outros_nomes = nomes.copy()

outros_nomes[0] = "Carlos"

print(nomes)
print(outros_nomes)
```

Nesse caso, `nomes` e `outros_nomes` são listas diferentes.

## Loops

Um `loop` permite executar um bloco de código repetidamente.

### `for`

O `for` é utilizado para percorrer os elementos de uma estrutura, como uma lista.

```python
nomes = ["Gabriel", "Maria", "João"]

for nome in nomes:
    print(nome)
```

A variável `nome` recebe, a cada iteração, o próximo elemento da lista.

#### `range()`

A função `range()` pode ser utilizada para gerar uma sequência de números, sendo bastante útil em loops.

```python
for numero in range(5):
    print(numero)
```

Por padrão, a sequência começa em `0` e termina antes do valor informado.

Também podemos definir o início e o fim:

```python
for numero in range(2, 6):
    print(numero)
```

E podemos definir o passo:

```python
for numero in range(0, 10, 2):
    print(numero)
```

### `while`

O `while` executa um bloco de código enquanto uma condição for verdadeira.

```python
contador = 0

while contador < 5:
    print(contador)
    contador += 1
```

É importante garantir que a condição do `while` eventualmente se torne falsa. Caso contrário, teremos um loop infinito.

### `break`

O `break` interrompe imediatamente a execução do loop.

```python
for numero in range(10):
    if numero == 5:
        break

    print(numero)

numero = 0

while True:
    if numero == 5:
            break
    
    print(numero)
    numero = numero + 1
```
