## Dicionários

Um dicionário é uma estrutura utilizada para armazenar valores associados a chaves.

Cada elemento possui uma chave e um valor correspondente:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}
```

As chaves são utilizadas para identificar e acessar os valores armazenados no dicionário.

### Acessando elementos

Para acessar um valor, utilizamos sua chave:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

print(aluno["nome"])
print(aluno["idade"])
```

Também podemos utilizar o método `get()`:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

print(aluno.get("nome"))
```

Uma diferença importante é que o `get()` retorna `None` caso a chave não exista, enquanto o acesso direto utilizando `[]` gera um `KeyError`.

```python
aluno = {
    "nome": "Gabriel"
}

print(aluno.get("email"))
```

### Listas dentro de dicionários

Os valores de um dicionário podem ser qualquer objeto, inclusive listas.

```python
aluno = {
    "nome": "Gabriel",
    "notas": [8.5, 7.0, 9.0]
}

print(aluno["notas"])
print(aluno["notas"][0])
```

Nesse caso, `aluno["notas"]` retorna uma lista, e podemos utilizar seu índice para acessar um elemento específico.

### Dicionários dentro de dicionários

Um dicionário também pode conter outros dicionários como valores.

```python
aluno = {
    "nome": "Gabriel",
    "endereco": {
        "cidade": "Londrina",
        "estado": "Paraná"
    }
}

print(aluno["endereco"])
print(aluno["endereco"]["cidade"])
```

Podemos combinar diferentes estruturas:

```python
aluno = {
    "nome": "Gabriel",
    "notas": [8.5, 7.0, 9.0],
    "endereco": {
        "cidade": "Londrina",
        "estado": "Paraná"
    }
}

print(aluno["notas"][1])
print(aluno["endereco"]["estado"])
```

### Alterando valores

Os valores associados às chaves podem ser alterados após a criação do dicionário.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

aluno["idade"] = 26

print(aluno)
```

Também podemos alterar valores de estruturas que estão dentro do dicionário:

```python
aluno = {
    "nome": "Gabriel",
    "notas": [8.5, 7.0, 9.0]
}

aluno["notas"][1] = 8.0

print(aluno)
```

### Adicionando uma nova chave

Podemos adicionar uma nova chave simplesmente atribuindo um valor a uma chave que ainda não existe.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

aluno["curso"] = "Computação"

print(aluno)
```

Também podemos adicionar estruturas inteiras como valores:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

aluno["notas"] = [8.5, 7.0, 9.0]

print(aluno)
```

### Removendo uma chave

O comando `del` pode ser utilizado para remover uma chave e seu respectivo valor.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

del aluno["idade"]

print(aluno)
```

O método `pop()` também remove uma chave, mas retorna o valor removido.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

curso = aluno.pop("curso")

print(curso)
print(aluno)
```

### Verificando o tamanho

A função `len()` retorna a quantidade de chaves existentes no dicionário.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

print(len(aluno))
```

### Percorrendo um dicionário

O `for` pode ser utilizado para percorrer as chaves de um dicionário.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

for chave in aluno:
    print(chave)
```

Podemos utilizar o método `items()` para obter a chave e o valor durante a iteração:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

for chave, valor in aluno.items():
    print(chave, valor)
```


## Tuplas

Uma tupla é uma estrutura utilizada para armazenar múltiplos valores, assim como uma lista.

A principal diferença é que tuplas são imutáveis. Depois de criada, não é possível alterar, adicionar ou remover seus elementos.

```python
coordenadas = (10, 20)

print(coordenadas)
```

Os elementos de uma tupla são acessados por meio de índices, assim como nas listas:

```python
coordenadas = (10, 20, 30)

print(coordenadas[0])
print(coordenadas[1])
print(coordenadas[-1])
```

### Tuplas são imutáveis

Não podemos alterar um elemento de uma tupla:

```python
coordenadas = (10, 20, 30)

coordenadas[0] = 100
```

A tentativa de alteração gera um `TypeError`.

### Criando uma tupla com um único elemento

Para criar uma tupla com apenas um elemento, é necessário utilizar uma vírgula:

```python
numero = (10,)

print(type(numero))
```

Sem a vírgula, os parênteses apenas agrupam a expressão:

```python
numero = (10)

print(type(numero))
```

### Tuplas podem conter diferentes objetos

Assim como listas e dicionários, tuplas podem armazenar objetos de diferentes tipos.

```python
dados = ("Gabriel", 25, 1.75, True)

print(dados)
```

Também podem conter outras estruturas:

```python
aluno = (
    "Gabriel",
    [8.5, 7.0, 9.0],
    {
        "cidade": "Londrina",
        "estado": "Paraná"
    }
)

print(aluno[0])
print(aluno[1][0])
print(aluno[2]["cidade"])
```

Embora a tupla seja imutável, os objetos armazenados nela podem ser mutáveis. Portanto, uma lista dentro de uma tupla pode ser alterada:

```python
dados = ("Gabriel", [8.5, 7.0, 9.0])

dados[1][0] = 10.0

print(dados)
```

### Slicing em tuplas

Assim como nas listas, podemos utilizar *slicing* para obter uma parte da tupla.

```python
numeros = (0, 1, 2, 3, 4, 5)

print(numeros[1:4])
print(numeros[::-1])
```

### Percorrendo uma tupla

Uma tupla pode ser percorrida utilizando um `for`.

```python
nomes = ("Gabriel", "Maria", "João")

for nome in nomes:
    print(nome)
```

## Quando utilizar listas ou tuplas?

De forma geral:

* Lista: quando os elementos podem ser alterados durante a execução do programa.
* Tupla: quando os elementos devem permanecer imutáveis após sua criação.
