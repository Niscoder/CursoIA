## Operador `in`

O operador `in` verifica se um elemento está presente em uma estrutura.

Em uma lista, podemos utilizá-lo para verificar se um determinado valor está presente:

```python
nomes = ["Gabriel", "Maria", "João"]

print("Maria" in nomes)
print("Carlos" in nomes)
```

O resultado será um valor booleano (`True` ou `False`).

Também podemos utilizar o operador `not in` para verificar se um elemento não está presente:

```python
nomes = ["Gabriel", "Maria", "João"]

print("Carlos" not in nomes)
```

O operador `in` pode ser utilizado diretamente em uma condição:

```python
nomes = ["Gabriel", "Maria", "João"]

nome = "Maria"

if nome in nomes:
    print("Nome encontrado")
else:
    print("Nome não encontrado")
```

### `in` em dicionários

Ao utilizar `in` diretamente em um dicionário, a busca é realizada nas chaves.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

print("nome" in aluno)
print("email" in aluno)
```

Também podemos utilizar `in` para verificar se uma chave existe antes de acessá-la:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

if "email" in aluno:
    print(aluno["email"])
else:
    print("E-mail não cadastrado")
```

Para verificar valores de um dicionário, podemos utilizar `values()`:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25
}

print("Gabriel" in aluno.values())
print(30 in aluno.values())
```

### `enumerate()`

A função `enumerate()` permite percorrer uma estrutura obtendo o índice e o valor de cada elemento.

```python
nomes = ["Gabriel", "Maria", "João"]

for indice, nome in enumerate(nomes):
    print(indice, nome)
```

Também podemos definir o valor inicial do índice:

```python
nomes = ["Gabriel", "Maria", "João"]

for indice, nome in enumerate(nomes, start=1):
    print(indice, nome)
```

## `continue`

A instrução `continue` interrompe a iteração atual do loop e passa diretamente para a próxima iteração.

```python
for numero in range(5):
    if numero == 2:
        continue

    print(numero)
```

Nesse exemplo, quando `numero` é igual a `2`, o `continue` faz com que o restante do bloco seja ignorado naquela iteração.

Podemos utilizar essa instrução para ignorar determinados elementos durante o processamento:

```python
numeros = [1, 2, 3, 4, 5]

for numero in numeros:
    if numero % 2 == 0:
        continue

    print(numero)
```

Nesse caso, os números pares são ignorados e somente os números ímpares são processados.

## Percorrendo as chaves de um dicionário

O método `keys()` retorna as chaves de um dicionário.

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

for chave in aluno.keys():
    print(chave)
```

Como percorrer as chaves é o comportamento padrão de um `for` em dicionários, o uso de `keys()` é opcional nesse caso:

```python
for chave in aluno:
    print(chave)
```

## Percorrendo os valores de um dicionário

O método `values()` permite percorrer somente os valores:

```python
aluno = {
    "nome": "Gabriel",
    "idade": 25,
    "curso": "Computação"
}

for valor in aluno.values():
    print(valor)
```
