## Strings

Strings são objetos utilizados para representar textos.

Em Python, strings podem ser delimitadas por aspas simples (`'`) ou aspas duplas (`"`).

```python
nome = 'Pedro'
cidade = "Londrina"

print(nome)
print(cidade)
```

As duas formas possuem o mesmo comportamento. A escolha normalmente depende do contexto ou de qual tipo de aspas facilita a escrita do texto.

Por exemplo, podemos utilizar aspas duplas quando o texto contém aspas simples:

```python
mensagem = "O aluno disse 'Olá!'"

print(mensagem)
```

Também podemos utilizar aspas simples quando o texto contém aspas duplas:

```python
mensagem = 'O professor disse "Olá!"'

print(mensagem)
```

### Strings como sequência

Uma string pode ser entendida como uma sequência de caracteres. Assim como em uma lista, podemos acessar seus elementos utilizando índices.

```python
nome = "Python"

print(nome[0])
print(nome[1])
print(nome[-1])
```

Os índices começam em `0` e índices negativos permitem acessar os caracteres a partir do final.

### Slicing de strings

Também podemos utilizar *slicing* para obter uma parte da string.

```python
texto = "Python"

print(texto[0:3])
print(texto[2:])
print(texto[:4])
```

Assim como nas listas, podemos utilizar um `step`:

```python
texto = "Python"

print(texto[::2])
print(texto[::-1])
```

### Strings são imutáveis

Strings são objetos imutáveis. Isso significa que não podemos alterar diretamente um caractere utilizando seu índice.

```python
texto = "Python"

texto[0] = "J"
```

A tentativa de alteração gera um `TypeError`.

Para obter uma string diferente, precisamos criar um novo objeto:

```python
texto = "Python"

texto = "J" + texto[1:]

print(texto)
```

### F-strings

Strings podem utilizar o prefixo `f` para inserir valores e expressões diretamente no texto.

```python
nome = "Pedro"
idade = 25

mensagem = f"Meu nome é {nome} e tenho {idade} anos."

print(mensagem)
```

Dentro das chaves `{}` podemos utilizar expressões e realizar operações:

```python
a = 10
b = 5

print(f"A soma é {a + b}")
print(f"A multiplicação é {a * b}")
```

Também podemos utilizar métodos e outras expressões:

```python
nome = "Pedro"

print(f"Nome em maiúsculas: {nome.upper()}")
```

### Comparando strings

Os operadores de comparação podem ser utilizados para comparar strings.

```python
nome1 = "Pedro"
nome2 = "Pedro"

print(nome1 == nome2)
print(nome1 != nome2)
```

O operador `==` verifica se as strings possuem o mesmo conteúdo, enquanto `!=` verifica se possuem conteúdos diferentes.

Também podemos utilizar `>`, `<`, `>=` e `<=`. Nesse caso, a comparação é feita seguindo a ordem dos caracteres definida pelo Unicode.

```python
print("abc" < "abd")
print("Python" == "Python")
print("Python" != "python")
```

A comparação é **case-sensitive**, portanto letras maiúsculas e minúsculas são consideradas diferentes.

### Tamanho da string

A função `len()` retorna a quantidade de caracteres da string.

```python
texto = "Python"

print(len(texto))
```

### Principais métodos de strings

A classe `str` possui diversos métodos para manipulação e consulta de textos.

#### `upper()`

Converte todos os caracteres para maiúsculo.

```python
texto = "Python"

print(texto.upper())
```

#### `lower()`

Converte todos os caracteres para minúsculo.

```python
texto = "Python"

print(texto.lower())
```

#### `capitalize()`

Converte o primeiro caractere para maiúsculo e os demais para minúsculo.

```python
texto = "pYTHON"

print(texto.capitalize())
```

#### `title()`

Converte a primeira letra de cada palavra para maiúsculo.

```python
texto = "python para ciência de dados"

print(texto.title())
```

#### `strip()`

Remove espaços em branco no início e no final da string.

```python
texto = "   Python   "

print(texto.strip())
```

Também existem `lstrip()` e `rstrip()`, que removem espaços somente do início ou do final, respectivamente.

#### `replace()`

Substitui uma parte da string por outra.

```python
texto = "Eu gosto de Java"

texto = texto.replace("Java", "Python")

print(texto)
```

#### `split()`

Divide uma string em partes e retorna uma lista.

```python
texto = "Python Java C"

linguagens = texto.split()

print(linguagens)
```

Podemos informar o separador:

```python
texto = "Python,Java,C"

linguagens = texto.split(",")

print(linguagens)
```

#### `join()`

Une elementos de uma sequência utilizando a string como separador.

```python
linguagens = ["Python", "Java", "C"]

texto = ", ".join(linguagens)

print(texto)
```

#### `find()`

Retorna o índice da primeira ocorrência de um texto.

```python
texto = "Eu gosto de Python"

print(texto.find("Python"))
print(texto.find("Java"))
```

Quando o texto não é encontrado, `find()` retorna `-1`.

#### `count()`

Retorna a quantidade de ocorrências de um determinado texto.

```python
texto = "Python é uma linguagem. Python é muito utilizada."

print(texto.count("Python"))
```

#### `startswith()`

Verifica se a string começa com determinado texto.

```python
texto = "Python é uma linguagem"

print(texto.startswith("Python"))
print(texto.startswith("Java"))
```

#### `endswith()`

Verifica se a string termina com determinado texto.

```python
arquivo = "dados.csv"

print(arquivo.endswith(".csv"))
print(arquivo.endswith(".txt"))
```

#### `isdigit()`

Verifica se todos os caracteres da string são dígitos.

```python
valor = "12345"

print(valor.isdigit())
```

#### `isalpha()`

Verifica se todos os caracteres da string são letras.

```python
texto = "Python"

print(texto.isalpha())
```

#### `isalnum()`

Verifica se todos os caracteres da string são letras ou números.

```python
texto = "Python123"

print(texto.isalnum())
```

#### `removeprefix()`

Remove o prefixo informado somente se ele estiver no início da string.

```python
texto = "Python é uma linguagem"

print(texto.removeprefix("Python"))
```

#### `removesuffix()`

Remove o sufixo informado somente se ele estiver no final da string.

```python
arquivo = "documento.pdf"

print(arquivo.removesuffix(".pdf"))
```

## Strings e loops

Como strings são sequências de caracteres, também podemos percorrê-las utilizando um `for`.

```python
texto = "Python"

for caractere in texto:
    print(caractere)
```

Isso permite processar cada caractere individualmente.
