## Funções

Funções são blocos de código que agrupam uma determinada tarefa e podem ser executados várias vezes ao longo do programa.

Elas ajudam a:

* evitar repetição de código;
* organizar o programa;
* facilitar a manutenção;
* reutilizar uma mesma lógica em diferentes partes do código.

### Criando uma função

Uma função é definida utilizando a palavra-chave `def`:

```python
def saudacao():
    print("Olá!")
```

Para executar a função, basta chamá-la pelo nome:

```python
saudacao()
```

### Retorno de uma função

Uma função pode produzir um resultado utilizando `return`:

```python
def soma():
    return 10 + 20
```

O valor retornado pode ser armazenado em uma variável:

```python
resultado = soma()

print(resultado)
```

O `return` também encerra a execução da função naquele ponto.

```python
def verificar_idade():
    idade = 20

    if idade >= 18:
        return "Maior de idade"

    return "Menor de idade"
```

### Parâmetros

Parâmetros permitem que uma função receba valores para utilizar durante sua execução.

```python
def saudacao(nome):
    print(f"Olá, {nome}!")
```

Ao chamar a função, fornecemos o valor do parâmetro:

```python
saudacao("João")
saudacao("Maria")
```

Nesse caso, `"João"` e `"Maria"` são argumentos passados para o parâmetro `nome`.

Podemos utilizar vários parâmetros:

```python
def soma(a, b):
    return a + b
```

```python
resultado = soma(10, 20)

print(resultado)
```

### Parâmetros posicionais

Por padrão, os argumentos são associados aos parâmetros pela posição em que são informados.


```python
def apresentar(nome, idade):
    print(f"Nome: {nome}")
    print(f"Idade: {idade}")

apresentar("João", 25)
```

A ordem dos argumentos é importante:

```python
apresentar(25, "João")
```

Python não impede que os tipos sejam diferentes dos esperados pela função. A consequência dependerá de como os parâmetros serão utilizados dentro dela.

### Argumentos nomeados

Também podemos informar explicitamente o nome do parâmetro:

```python
apresentar(nome="João", idade=25)
```

Nesse caso, a associação não depende da ordem:

```python
apresentar(idade=25, nome="João")
```

Podemos combinar argumentos posicionais e nomeados, desde que os posicionais apareçam primeiro:

```python
apresentar("João", idade=25)
```

### Parâmetros com valores padrão

Podemos definir um valor padrão para um parâmetro.

```python
def saudacao(nome, mensagem="Olá"):
    print(f"{mensagem}, {nome}!")
```

Se o argumento não for informado, o valor padrão será utilizado:

```python
saudacao("João")

saudacao("João", "Bom dia")
```

Parâmetros com valores padrão são opcionais durante a chamada da função.

Parâmetros obrigatórios devem aparecer antes dos parâmetros que possuem valores padrão:

```python
def exemplo(nome, idade=18):
    print(nome, idade)
```

O seguinte não é permitido:

```python
def exemplo(idade=18, nome):
    print(nome, idade)
```

### `*args`

Em algumas situações, não sabemos antecipadamente quantos argumentos uma função receberá.

O parâmetro `*args` permite receber uma quantidade variável de argumentos posicionais.

```python
def soma(*args):
    print(args)

soma(10, 20, 30)
```

Os argumentos recebidos por `*args` são armazenados em uma tupla.

Podemos percorrer os valores normalmente:

```python
def soma(*args):
    resultado = 0

    for valor in args:
        resultado += valor

    return resultado

print(soma(10, 20))
print(soma(10, 20, 30, 40))
```

Também é possível não passar nenhum argumento:

```python
print(soma())
```

### `**kwargs`

Assim como `*args` permite receber vários argumentos posicionais, `**kwargs` permite receber vários argumentos nomeados.

```python
def apresentar(**kwargs):
    print(kwargs)

apresentar(nome="João", idade=25, cidade="Londrina")
```

Os argumentos recebidos por `**kwargs` são armazenados em um dicionário.

Podemos percorrer suas chaves e valores:

```python
def apresentar(**kwargs):
    for chave, valor in kwargs.items():
        print(f"{chave}: {valor}")

apresentar(nome="João", idade=25, cidade="Londrina")
```

### Utilizando `*args` e `**kwargs` juntos

Uma função pode receber tanto argumentos posicionais variáveis quanto argumentos nomeados variáveis:

```python
def exemplo(*args, **kwargs):
    print("args:", args)
    print("**kwargs:", kwargs)

exemplo(10, 20, 30, nome="João", idade=25)
```

Nesse caso:

* `args` contém os argumentos posicionais;
* `**kwargs` contém os argumentos nomeados.

### Combinando parâmetros

Podemos combinar parâmetros obrigatórios, parâmetros com valor padrão, `*args` e `**kwargs`.

```python
def exemplo(nome, idade=18, *args, **kwargs):
    print("nome:", nome)
    print("idade:", idade)
    print("args:", args)
    print("**kwargs:", kwargs)

exemplo(
    "João",
    25,
    "Python",
    "Machine Learning",
    cidade="Londrina",
    curso="TI"
)
```

### Escopo de variáveis

Variáveis criadas dentro de uma função normalmente pertencem ao escopo daquela função.

A variável `mensagem` pode ser utilizada dentro da função, mas não está disponível fora dela:

```python
def exemplo():
    mensagem = "Olá!"
    print(mensagem)

print(mensagem)
```

Variáveis definidas fora da função pertencem a um escopo externo e podem ser lidas dentro da função:

```python
mensagem = "Olá!"

def exemplo():
    print(mensagem)

exemplo()
```
