
# Análise Exploratória de Dados com Seaborn

Material sobre **Seaborn**, em conjunto com Pandas e Matplotlib, para explorar um conjunto de dados.

> Seaborn é principalmente uma biblioteca de visualização estatística. Os cálculos de média, covariância e correlação continuam sendo realizados pelo Pandas/NumPy.


## Importação das bibliotecas

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (9, 5)
```

## Criação do conjunto de dados

```python

dados = {
    "nome": [
        "Ana", "Bruno", "Carlos", "Daniela", "Eduardo",
        "Gustavo", "Gabriel", "Marcos", "Fernanda", "Juliana",
        "Lucas", "Patrícia", "Rafael", "Sofia", "Thiago",
        "Vanessa", "Aline", "Beatriz", "Caio", "Diego"
    ],
    "idade": [
        22, 25, 21, 28, 24, 23, 22, 20, 27, 26,
        23, 29, 24, 21, 30, 22, 25, 27, 20, 31
    ],
    "horas_estudo": [
        8, 12, 15, 5, 10, 14, 7, 16, 9, 11,
        13, 4, 10, 17, 3, 8, 12, 6, 18, 5
    ],
    "curso": [
        "Python", "Python", "Machine Learning", "Python",
        "Machine Learning", "Machine Learning", "Python",
        "Machine Learning", "Python", "Machine Learning",
        "Python", "Machine Learning", "Python", "Machine Learning",
        "Python", "Machine Learning", "Python", "Machine Learning",
        "Python", "Machine Learning"
    ],
    "nota": [
        8.5, 7.0, 9.2, 6.5, 8.0, 9.0, 7.4, 8.8, 8.1, 7.8,
        8.7, 5.9, 7.5, 9.5, 5.2, 7.2, 8.9, 6.8, 9.1, 6.1
    ]
}

df = pd.DataFrame(dados)
df.head()
```

```python
df.info()
```


## Distribuição dos dados

Uma distribuição mostra como os valores de uma variável estão espalhados.

Perguntas importantes:

- Os valores estão concentrados em uma região?
- A distribuição parece simétrica?
- Existem valores extremos?
- A distribuição possui mais de um pico?
- A distribuição muda entre categorias?


### Histograma

```python

sns.histplot(data=df, x="nota", bins=8)

plt.title("Distribuição das notas")
plt.xlabel("Nota")
plt.ylabel("Quantidade de alunos")
plt.show()
```


### Histograma com KDE

O parâmetro `kde=True` adiciona uma estimativa suavizada da distribuição. A curva ajuda a observar a forma geral dos dados, mas não representa uma contagem exata.


```python

sns.histplot(data=df, x="nota", bins=8, kde=True)

plt.title("Distribuição das notas com KDE")
plt.xlabel("Nota")
plt.ylabel("Quantidade / densidade")
plt.show()
```

### Distribuições separadas por categoria

```python

sns.histplot(
    data=df,
    x="nota",
    hue="curso",
    bins=8,
    kde=True,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Distribuição das notas por curso")
plt.xlabel("Nota")
plt.ylabel("Densidade")
plt.show()
```

### Distribuição das horas de estudo

```python

sns.histplot(data=df, x="horas_estudo", bins=8, kde=True)

plt.title("Distribuição das horas de estudo")
plt.xlabel("Horas de estudo")
plt.ylabel("Quantidade")
plt.show()
```


## Boxplots

O boxplot resume uma distribuição usando:

- Primeiro quartil — Q1;
- Mediana — Q2;
- Terceiro quartil — Q3;
- Intervalo interquartil — IQR;
- Possíveis valores extremos.

Por convenção, os boxplots consideram como possíveis outliers os valores abaixo de `Q1 - 1.5 × IQR` ou acima de `Q3 + 1.5 × IQR`.


### Boxplot geral

```python

sns.boxplot(data=df, y="nota")

plt.title("Boxplot das notas")
plt.ylabel("Nota")
plt.show()
```

### Boxplot por curso

```python

sns.boxplot(data=df, x="curso", y="nota")

plt.title("Distribuição das notas por curso")
plt.xlabel("Curso")
plt.ylabel("Nota")
plt.xticks(rotation=15)
plt.show()
```

### Violinplot

```python

sns.violinplot(data=df, x="curso", y="nota", inner="box")

plt.title("Violinplot das notas por curso")
plt.xlabel("Curso")
plt.ylabel("Nota")
plt.xticks(rotation=15)
plt.show()
```


O violinplot combina características do boxplot com uma estimativa da densidade. Com poucos dados, sua forma deve ser interpretada com cuidado.



## Relações entre variáveis

Agora vamos investigar se existe algum padrão entre duas variáveis numéricas.

O scatterplot representa cada observação como um ponto.


### Scatterplot

```python

sns.scatterplot(data=df, x="horas_estudo", y="nota")

plt.title("Relação entre horas de estudo e nota")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.show()
```

### Scatterplot usando `hue`

```python

sns.scatterplot(
    data=df,
    x="horas_estudo",
    y="nota",
    hue="curso",
    s=100
)

plt.title("Horas de estudo versus nota por curso")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.show()
```

### Linha de tendência com `regplot`

```python

sns.regplot(
    data=df,
    x="horas_estudo",
    y="nota",
    scatter_kws={"s": 60},
    line_kws={"color": "red"}
)

plt.title("Tendência linear entre estudo e nota")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.show()
```

### Pairplot

```python

sns.pairplot(
    df,
    vars=["idade", "horas_estudo", "nota"],
    hue="curso"
)

plt.show()
```


## Covariância

A covariância indica se duas variáveis tendem a variar juntas:

- Positiva: quando uma aumenta, a outra tende a aumentar;
- Negativa: quando uma aumenta, a outra tende a diminuir;
- Próxima de zero: pouca evidência de relação linear.

A covariância depende das unidades das variáveis.


```python

covariancia = df[["idade", "horas_estudo", "nota"]].cov()
covariancia
```

### Heatmap da covariância

```python

plt.figure(figsize=(8, 6))

sns.heatmap(
    covariancia,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Matriz de covariância")
plt.show()
```


## Correlação

A correlação de Pearson mede a intensidade e a direção de uma relação linear.

- `1`: relação linear positiva perfeita;
- `-1`: relação linear negativa perfeita;
- `0`: ausência de relação linear.

A correlação é padronizada entre -1 e 1, sendo mais fácil de comparar que a covariância.

> **Importante:** correlação não implica causalidade.


```python

correlacao = df[["idade", "horas_estudo", "nota"]].corr()
correlacao
```

### Heatmap da correlação

```python

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlacao,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    square=True
)

plt.title("Matriz de correlação")
plt.show()
```


### Comparando covariância e correlação

| Medida | Característica |
|---|---|
| Covariância | Depende das unidades das variáveis |
| Correlação | Padronizada entre -1 e 1 |
| Covariância | Indica direção de variação conjunta |
| Correlação | Indica direção e intensidade da relação linear |

Uma covariância alta não significa necessariamente uma correlação alta, pois a covariância é influenciada pela escala das variáveis.
