# Introdução às redes neurais artificiais

As redes neurais artificiais são modelos de aprendizado de máquina compostos por unidades de processamento conectadas, capazes de combinar informações e aprender padrões a partir de exemplos.

## O que é um perceptron?

O perceptron é um dos modelos mais simples de neurônio artificial. Ele recebe valores de entrada, associa um peso a cada entrada, combina esses valores e produz uma saída.

Na forma clássica, utiliza uma função de ativação limiar: a saída assume uma de duas classes conforme a soma ponderada ultrapassa ou não determinado limite. Por isso, pode ser usado em tarefas simples de classificação binária.

O perceptron foi um marco no desenvolvimento das redes neurais, mas possui limitações: um perceptron simples só aprende fronteiras de decisão lineares. Problemas que não podem ser separados linearmente, como o XOR, não podem ser resolvidos por um único perceptron clássico.

## Como funciona um neurônio artificial?

Um neurônio artificial combina as entradas com seus respectivos pesos e adiciona um termo chamado **viés** (*bias*). Em seguida, aplica uma função de ativação ao resultado.

$
z = \sum_{i=1}^{n} w_i x_i + b
$

$
\hat{y} = f(z)
$

Em que:

- $x_i$: valores de entrada, ou características;
- $w_i$: pesos associados às entradas;
- $b$: viés, que desloca a combinação ponderada;
- $z$: soma ponderada antes da ativação;
- $f$: função de ativação;
- $\hat{y}$: saída produzida pelo neurônio.

Os **pesos** determinam a influência relativa de cada entrada. O **viés** permite ajustar o ponto em que o neurônio responde. A função de ativação transforma a combinação ponderada na saída do neurônio.

## O papel da função de ativação

A função de ativação define como a soma ponderada será convertida em uma saída. No perceptron clássico, utiliza-se uma função limiar que produz uma decisão discreta. Nas redes neurais modernas, são comuns funções como ReLU, sigmoide e tangente hiperbólica, dependendo da arquitetura e do problema.

A escolha da função de ativação influencia as relações que a rede consegue representar e como seus parâmetros podem ser ajustados durante o treinamento.

Um ponto fundamental é a **não linearidade**. Sem funções de ativação não lineares entre as camadas, uma sequência de transformações lineares equivale, na prática, a uma única transformação linear. Adicionar camadas, por si só, não permitiria representar fronteiras de decisão não lineares mais complexas.

## Do perceptron simples à rede neural

Um perceptron simples possui uma única unidade de decisão e capacidade limitada para representar padrões. Uma rede neural combina muitos neurônios artificiais, organizados em camadas e conectados entre si.

Em uma arquitetura comum, encontramos:

- **Camada de entrada:** recebe as características do exemplo e encaminha os valores.
- **Camadas ocultas:** combinam as saídas anteriores usando pesos, vieses e funções de ativação. Podem aprender representações intermediárias.
- **Camada de saída:** produz a resposta final, como uma classe prevista ou um valor numérico.

A quantidade de camadas, de neurônios em cada camada e as funções de ativação são escolhas de arquitetura. Redes mais complexas podem representar padrões mais elaborados, mas também exigem mais dados, recursos computacionais e cuidado para evitar sobreajuste.

## Rede perceptron multicamadas (MLP)

A **Multilayer Perceptron (MLP)**, ou rede perceptron multicamadas, é uma rede neural formada por uma camada de entrada, uma ou mais camadas ocultas e uma camada de saída. Os neurônios de uma camada enviam seus resultados para a camada seguinte, conforme as conexões definidas na arquitetura.

Apesar do nome, uma MLP moderna normalmente utiliza funções de ativação não lineares nas camadas ocultas, o que permite aprender relações mais complexas do que um perceptron simples.

Características importantes:

- possui várias camadas de processamento;
- aprende pesos e vieses a partir dos dados;
- pode ser usada em classificação e regressão;
- consegue representar relações não lineares;
- seu desempenho depende da arquitetura, dos dados, das funções de ativação e do processo de treinamento.

## O que significa dizer que uma rede é feedforward?

Uma rede **feedforward** é aquela em que a informação percorre a rede em uma única direção: da entrada, passando pelas camadas ocultas, até a saída. Não há ciclos em que a saída de uma camada retorne a uma camada anterior como parte do fluxo de inferência.

Em uma MLP feedforward, cada camada recebe os valores da camada anterior, calcula combinações ponderadas e ativações, e encaminha os resultados para a próxima camada.

É importante distinguir:

- **Propagação direta (*forward pass*):** calcula a previsão usando os pesos e vieses atuais.
- **Treinamento:** compara a previsão com a resposta esperada e utiliza essa diferença para ajustar os parâmetros.

Uma rede pode ser feedforward na arquitetura e utilizar backpropagation durante o treinamento. Feedforward descreve o fluxo de informação; backpropagation é um método para calcular como a perda depende dos parâmetros.

## Como a rede aprende: função de perda

Para treinar uma rede com exemplos rotulados, precisamos medir o quanto suas previsões diferem das respostas esperadas. Isso é feito por uma **função de perda** (*loss function*).

A função de perda resume o erro do modelo em uma ou mais observações. A escolha depende da tarefa: classificação e regressão costumam utilizar funções diferentes.

O objetivo é encontrar pesos e vieses que reduzam a perda nos dados de treino e permitam bom desempenho em exemplos ainda não vistos. Uma perda baixa no treino, por si só, não garante boa generalização: pode ocorrer **sobreajuste** (*overfitting*).

## O que é backpropagation?

O **backpropagation** (*retropropagação do erro*) é um procedimento para calcular os gradientes da função de perda em relação aos pesos e vieses da rede. Ele aplica a regra da cadeia do cálculo diferencial, propagando essas informações da saída em direção às camadas anteriores.

De forma intuitiva:

1. A rede realiza a propagação direta e produz uma previsão.
2. A função de perda compara a previsão com a resposta esperada.
3. O backpropagation calcula os gradientes, indicando como a perda varia em relação a cada peso e viés.
4. Um algoritmo de otimização utiliza os gradientes para atualizar os parâmetros.
5. O processo se repete ao longo dos exemplos e das épocas de treinamento.

**Backpropagation não é, por si só, o algoritmo que atualiza os pesos.** Ele calcula os gradientes necessários para a atualização. O otimizador — por exemplo, gradiente descendente ou Adam — usa esses gradientes para modificar os parâmetros.

## Como os pesos são ajustados?

Após calcular os gradientes, o otimizador ajusta pesos e vieses para tentar reduzir a função de perda. Uma regra simples, conhecida como **gradiente descendente**, pode ser expressa por:

$
w_{\text{novo}} = w_{\text{atual}} - \eta \frac{\partial L}{\partial w}
$

Em que:

- $w$: um peso da rede;
- $L$: função de perda;
- $\frac{\partial L}{\partial w}$: gradiente da perda em relação ao peso;
- $\eta$: taxa de aprendizagem (*learning rate*).

A taxa de aprendizagem controla o tamanho dos passos. Se for muito alta, o treinamento pode oscilar ou não convergir; se for muito baixa, o aprendizado pode ser lento.

O mesmo princípio se aplica aos vieses. Em redes com muitos parâmetros, o processo é repetido iterativamente, usando gradientes calculados em uma observação ou em pequenos lotes.

## Épocas, lotes e treinamento

Termos comuns no treinamento de redes neurais:

- **Época (*epoch*):** uma passagem completa por todos os exemplos do conjunto de treino.
- **Lote (*batch*):** conjunto de exemplos usado para calcular uma atualização dos parâmetros.
- **Batch size:** quantidade de exemplos em cada lote.
- **Iteração:** uma etapa de atualização; a quantidade por época depende do tamanho dos lotes.
- **Taxa de aprendizagem:** controla o tamanho das atualizações realizadas pelo otimizador.

O treinamento costuma envolver muitas atualizações. Acompanhar a perda e métricas em dados de validação ajuda a verificar se a rede está aprendendo padrões que generalizam, em vez de apenas se ajustar aos dados de treino.

## Como os conceitos se conectam

Podemos resumir o ciclo de treinamento de uma MLP assim:

**Entradas → Propagação direta → Previsão → Função de perda → Backpropagation → Otimizador atualiza pesos e vieses → Nova iteração**

A propagação direta calcula a saída; a função de perda mede o erro; o backpropagation calcula os gradientes; e o otimizador usa esses gradientes para ajustar os parâmetros. Esses passos se repetem até que um critério de parada seja alcançado ou o treinamento seja interrompido.

## Resumo dos conceitos

| Conceito | Ideia principal |
|---|---|
| Perceptron | Modelo simples de neurônio artificial, com decisão linear na forma clássica |
| Neurônio artificial | Combina entradas, pesos e viés, e aplica uma função de ativação |
| MLP | Rede com camadas de entrada, ocultas e saída |
| Feedforward | Fluxo de informação da entrada para a saída, sem ciclos na arquitetura |
| Função de perda | Mede a diferença entre a previsão e a resposta esperada |
| Backpropagation | Calcula gradientes da perda em relação aos parâmetros |
| Otimizador | Usa gradientes para ajustar pesos e vieses |
| Taxa de aprendizagem | Controla o tamanho dos passos de atualização |
| Época | Uma passagem completa pelo conjunto de treino |
