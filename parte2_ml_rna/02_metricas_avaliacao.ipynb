# Métricas para avaliação de modelos

Após o treinamento de um modelo de Machine Learning, é necessário avaliar seu desempenho. As métricas de avaliação permitem comparar as previsões realizadas pelo modelo com os valores esperados ou analisar a qualidade das classes previstas.

A métrica adequada depende do tipo de tarefa. Para **regressão**, são utilizadas métricas que medem a diferença entre valores numéricos previstos e observados. Para **classificação**, são utilizadas métricas que avaliam as classes previstas pelo modelo.

## Avaliação de modelos de regressão

Em problemas de regressão, o modelo produz um valor numérico como saída. As métricas precisam, portanto, quantificar o quanto as previsões se afastam dos valores reais.

Entre as métricas mais utilizadas estão o **Erro Médio Quadrático (MSE)**, a **Raiz do Erro Médio Quadrático (RMSE)** e o **coeficiente de determinação ($R^2$)**.

## Erro Médio Quadrático (MSE)

O **Mean Squared Error (MSE)**, ou Erro Médio Quadrático, calcula a média dos quadrados dos erros entre os valores reais e os valores previstos pelo modelo.

O erro de cada previsão é calculado pela diferença entre o valor real e o valor previsto. Esse erro é elevado ao quadrado antes de ser incluído na média.

### Fórmula

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

Onde:

- $n$ é o número de observações;
- $y_i$ é o valor real da observação $i$;
- $\hat{y}_i$ é o valor previsto pelo modelo para a observação $i$.

Como os erros são elevados ao quadrado, erros maiores têm uma influência maior sobre o resultado. Quanto **menor** o MSE, melhor o desempenho do modelo.

### Exemplo conceitual do MSE

Considere três valores reais e suas respectivas previsões:

| Valor real | Previsão | Erro | Erro² |
|---:|---:|---:|---:|
| 10 | 8 | 2 | 4 |
| 20 | 21 | -1 | 1 |
| 30 | 33 | -3 | 9 |

O MSE é obtido pela média dos erros quadráticos:

$$
MSE = \frac{4 + 1 + 9}{3} = \frac{14}{3} \approx 4,67
$$

Portanto, quanto mais próximas estiverem as previsões dos valores reais, menor tende a ser o MSE.

## Raiz do Erro Médio Quadrático (RMSE)

O **Root Mean Squared Error (RMSE)**, ou Raiz do Erro Médio Quadrático, é obtido aplicando a raiz quadrada ao MSE.

### Fórmula

$$
RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}
$$

A principal vantagem do RMSE em relação ao MSE é que ele retorna o erro na **mesma unidade da variável que está sendo prevista**.

Por exemplo, se um modelo prevê preços em reais, o RMSE também será expresso em reais. Assim como no MSE, **quanto menor o RMSE, melhor** tende a ser o desempenho do modelo.

## Coeficiente de determinação (R²)

O **coeficiente de determinação ($R^2$)** é uma métrica que indica quanto da variabilidade da variável de saída é explicada pelo modelo.

Uma forma de calcular o $R^2$ é:

$$
R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}
{\sum_{i=1}^{n}(y_i - \bar{y})^2}
$$

Onde $\bar{y}$ representa a média dos valores reais.

De forma simplificada, o $R^2$ compara o erro produzido pelo modelo com a variação existente nos próprios dados. Um valor próximo de **1** indica que o modelo explica uma grande parte da variabilidade observada.

Em muitos problemas, valores de $R^2$ podem ser interpretados aproximadamente como uma proporção da variabilidade explicada pelo modelo. Por exemplo, um $R^2 = 0,80$ indica que o modelo explica cerca de 80% da variabilidade da variável de saída em relação à referência utilizada no cálculo.

É importante observar que o $R^2$ não deve ser interpretado isoladamente e que ele pode assumir valores negativos quando o modelo apresenta desempenho pior que uma referência baseada na média dos valores observados.

## Comparação das métricas de regressão

| Métrica | O que mede | Melhor resultado |
|---|---|---:|
| MSE | Média dos erros ao quadrado | Quanto menor, melhor |
| RMSE | Raiz da média dos erros ao quadrado | Quanto menor, melhor |
| $R^2$ | Proporção da variabilidade explicada pelo modelo | Geralmente, quanto maior, melhor |

MSE e RMSE são diretamente relacionados: o RMSE é simplesmente a raiz quadrada do MSE. O MSE dá maior peso a erros grandes, enquanto o RMSE possui a vantagem de estar na mesma unidade da variável prevista.

## Avaliação de modelos de classificação

Em problemas de classificação, o modelo atribui uma classe a cada exemplo. A avaliação compara as classes previstas com as classes reais.

Para classificação binária, os resultados podem ser organizados em uma **matriz de confusão**, formada por:

- **Verdadeiro Positivo (TP):** o modelo previu positivo e a classe real é positiva;
- **Verdadeiro Negativo (TN):** o modelo previu negativo e a classe real é negativa;
- **Falso Positivo (FP):** o modelo previu positivo, mas a classe real é negativa;
- **Falso Negativo (FN):** o modelo previu negativo, mas a classe real é positiva.

Esses resultados servem de base para diversas métricas, incluindo a **acurácia** e o **F1-score**.

## Acurácia

A **acurácia** representa a proporção de previsões que foram classificadas corretamente em relação ao total de previsões.

Para um problema de classificação binária, sua fórmula pode ser escrita como:

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

De forma geral:

$$
Accuracy = \frac{\text{número de previsões corretas}}
{\text{número total de previsões}}
$$

A acurácia é simples de interpretar, mas pode ser inadequada quando as classes estão muito desbalanceadas. Por exemplo, se 95% dos exemplos pertencem a uma única classe, um modelo que sempre escolha essa classe poderá alcançar 95% de acurácia sem necessariamente ser útil para identificar a classe minoritária.

## F1-score

O **F1-score** combina **precisão (precision)** e **revocação (recall)** em uma única métrica por meio da média harmônica.

A precisão mede, entre as amostras classificadas como positivas, quantas realmente são positivas:

$$
Precision = \frac{TP}{TP + FP}
$$

O recall mede, entre as amostras que realmente são positivas, quantas foram identificadas pelo modelo:

$$
Recall = \frac{TP}{TP + FN}
$$

O F1-score é calculado por:

$$
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

Também pode ser escrito diretamente em função de TP, FP e FN:

$$
F1 = \frac{2TP}{2TP + FP + FN}
$$

O valor do F1-score varia entre 0 e 1, sendo que valores maiores indicam melhor equilíbrio entre precisão e recall.

## F1-score micro

O **F1-score micro** calcula primeiro os totais de verdadeiros positivos, falsos positivos e falsos negativos considerando todas as classes em conjunto. A partir desses valores globais, são calculados precision, recall e F1.

De forma conceitual:

$$
Precision_{micro} = \frac{\sum TP}{\sum TP + \sum FP}
$$

$$
Recall_{micro} = \frac{\sum TP}{\sum TP + \sum FN}
$$

e então:

$$
F1_{micro} = 2 \cdot \frac{Precision_{micro} \cdot Recall_{micro}}
{Precision_{micro} + Recall_{micro}}
$$

Como todas as classes contribuem para os totais globais, o F1 micro tende a refletir mais fortemente o comportamento das classes que possuem maior quantidade de exemplos.

## F1-score macro

O **F1-score macro** calcula o F1-score individualmente para cada classe e, depois, calcula a média aritmética desses valores.

Para $K$ classes:

$$
F1_{macro} = \frac{1}{K}\sum_{k=1}^{K}F1_k
$$

Cada classe recebe o mesmo peso no resultado final, independentemente da quantidade de exemplos existentes naquela classe.

Por isso, o F1 macro é especialmente útil quando se deseja avaliar se o modelo apresenta um desempenho equilibrado entre as diferentes classes, inclusive quando existe desbalanceamento entre elas.

## Micro versus macro

| Métrica | Como é calculada | Influência das classes |
|---|---|---|
| F1 micro | Soma os resultados de todas as classes e calcula um único F1 | Classes maiores tendem a ter maior influência |
| F1 macro | Calcula um F1 para cada classe e depois faz a média | Todas as classes têm o mesmo peso |

Considere, por exemplo, uma base com três classes em que uma delas possui milhares de exemplos e outra possui poucos exemplos. O **F1 micro** será fortemente influenciado pela classe com maior quantidade de dados. Já o **F1 macro** atribuirá a mesma importância às três classes.

Assim, quando o objetivo é avaliar o desempenho geral considerando a distribuição dos exemplos, o F1 micro pode ser útil. Quando o objetivo é verificar se o modelo apresenta bom desempenho em todas as classes, o F1 macro pode ser mais informativo.

## Comparação das métricas de classificação

| Métrica | Característica | Melhor resultado |
|---|---|---:|
| Acurácia | Proporção de previsões corretas | Quanto maior, melhor |
| F1 micro | Equilíbrio global entre precisão e recall | Quanto maior, melhor |
| F1 macro | Média do F1 das classes, com peso igual | Quanto maior, melhor |

A escolha da métrica deve considerar o problema. Em bases balanceadas, a acurácia pode ser uma medida bastante útil. Em bases desbalanceadas, métricas como F1-score podem fornecer uma visão mais adequada do desempenho, especialmente quando é importante não ignorar classes minoritárias.
