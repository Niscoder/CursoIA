```python
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_iris.html
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_wine.html
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_diabetes.html

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.linear_model import SGDClassifier, LinearRegression
import numpy as np
```

```python
dataset = load_iris()

x = dataset.data
y = dataset.target
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = svm.SVC()
model.fit(X_treino, y_treino)
```

```python
model = SGDClassifier(loss="hinge", penalty="l2", max_iter=100)
model.fit(X_treino, y_treino)
```

```python
result = model.predict(X_teste)
```

```python
print(result)
```

```python
print(y_teste)
```

```python
from sklearn.metrics import confusion_matrix, f1_score, accuracy_score, root_mean_squared_error, mean_squared_error
```

```python
print(accuracy_score(y_teste, result))
print(f1_score(y_teste, result, average="micro"))
print(confusion_matrix(y_teste, result))
```

```python
rng = np.random.default_rng(seed=42)

x_train = rng.uniform(low=2000.0, high=15000.0, size=(500, 2))
x_test = rng.uniform(low=2000.0, high=15000.0, size=(50, 2))

y_train = rng.uniform(low=30.0, high=80.0, size=500)
y_test = rng.uniform(low=30.0, high=80.0, size=50)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
results = model.predict(x_test)
```

```python
print(results)
```

```python
print(mean_squared_error(y_test, results))
print(root_mean_squared_error(y_test, results))
```
