```python
import numpy as np
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.datasets import load_wine
from sklearn.pipeline import make_pipeline
```

```python
dataset = load_wine()
x = dataset.data
y = dataset.target
```

```python
models = [
    {"model": SVC(), "name": "SVC"},
    {"model": KNeighborsClassifier(), "name": "KNC"},
    {"model": SGDClassifier(), "name": "SGDC"}
]
```

```python
scalers = [
    {"scaler": MinMaxScaler(), "name": "MinMax"},
    {"scaler": StandardScaler(), "name": "Std"}
]
```

```python
feat_extrs = [
    {"pca": PCA(n_components=0.95), "name": "95%"},
    {"pca": PCA(n_components=0.88), "name": "88%"},
    {"pca": PCA(n_components=0.66), "name": "66%"},
    {"pca": PCA(n_components=0.50), "name": "50%"}
]
```

```python
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
for scaler in scalers:
    for feat_extr in feat_extrs:
        for model in models:
            print(f"Scaler: {scaler["name"]} - FeatExtr: {feat_extr["name"]} - Model: {model["name"]}")
            pipeline = make_pipeline(scaler["scaler"], feat_extr["pca"], model["model"])
            scores = cross_validate(pipeline, x, y, cv=kf, scoring={
                "accuracy": "accuracy",
                "f1_macro": "f1_macro"
            })
            print(f"acc - {np.mean(scores["test_accuracy"])}")
            print(f"f1score - {np.mean(scores["test_f1_macro"])}\n")
```
