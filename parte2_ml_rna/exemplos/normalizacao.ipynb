```python
from sklearn.datasets import load_wine
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import seaborn as sns
```

```python
dataset = load_wine()
```

```python
x = dataset.data
x_cols = dataset.feature_names

y = dataset.target
y_col = dataset.target_names
```

```python
print(x_cols)
```

```python
df = pd.DataFrame(data=x, columns=x_cols)
df["class"] = y
```

```python
df[x_cols].describe().round(2)
```

```python
samples = np.linspace(1, 179, 178)
plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=df["alcohol"], label="alcool", color="purple")
plt.scatter(x=samples, y=df["proline"], label="proline", color="green")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=df["alcohol"], label="alcool", color="purple")
# plt.scatter(x=samples, y=df["proline"], label="proline", color="green")
plt.legend()
plt.show()
```

```python
scaler = MinMaxScaler()
alcool_scaled = scaler.fit_transform(df[["alcohol"]])
plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=alcool_scaled, label="alcool", color="purple")
plt.legend()
plt.show()
```

```python
scaler = MinMaxScaler()
proline_scaled = scaler.fit_transform(df[["proline"]])
plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=proline_scaled, label="proline", color="green")
plt.legend()
plt.show()
```

```python
samples = np.linspace(1, 179, 178)
plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=alcool_scaled, label="alcool", color="purple")
plt.scatter(x=samples, y=proline_scaled, label="proline", color="green")
plt.legend()
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["alcohol", "proline", "malic_acid"],
    hue="class"
)

plt.show()
```

```python
scaler = StandardScaler()
alcool_scaled = scaler.fit_transform(df[["alcohol"]])

plt.figure(figsize=(10, 6))
plt.scatter(x=samples, y=alcool_scaled, label="alcool", color="purple")
plt.legend()
plt.show()
```

```python
df_alcool_scaled = pd.DataFrame(data=alcool_scaled, columns=["scaled_alcohol"])
df_alcool_scaled.describe().round(2)
```
