# Machine Learning

Material complementar: https://www.youtube.com/watch?v=E0Hmnixke2g

Machine Learning (ML), ou aprendizado de máquina, é uma área da Inteligência Artificial que busca desenvolver métodos capazes de aprender padrões a partir de dados para realizar previsões, classificações, agrupamentos ou tomadas de decisão.

Em vez de definir manualmente todas as regras de um problema, um algoritmo de Machine Learning utiliza dados para construir um modelo que pode ser aplicado a novos exemplos.

## Principais tipos de aprendizado

Os métodos de Machine Learning podem ser classificados de acordo com a forma como o algoritmo recebe informações durante o processo de aprendizagem. Os três grupos mais conhecidos são **aprendizado supervisionado**, **aprendizado não supervisionado** e **aprendizado por reforço**.

## Aprendizado supervisionado

No aprendizado supervisionado, o algoritmo recebe exemplos de entrada acompanhados das respostas esperadas. Essas respostas são chamadas de **rótulos** (*labels*).

O objetivo é aprender uma relação entre as características dos dados de entrada e seus respectivos rótulos, de modo que o modelo consiga produzir uma resposta adequada para dados que ainda não foram apresentados durante o treinamento.

Exemplos de aplicações:

- prever o preço de um imóvel a partir de suas características;
- identificar se uma mensagem é spam ou não;
- diagnosticar uma determinada condição a partir de características observadas;
- reconhecer categorias em imagens;
- prever a demanda de um produto.

### Exemplos de modelos supervisionados

Alguns algoritmos utilizados em aprendizado supervisionado são:

- **Regressão Linear**;
- **Regressão Logística**;
- **Árvores de Decisão**;
- **Random Forest**;
- **Support Vector Machine (SVM)**;
- **k-Nearest Neighbors (k-NN)**;
- **Redes Neurais Artificiais**.

A escolha do algoritmo depende das características do problema, dos dados disponíveis e do objetivo da tarefa.

## Aprendizado não supervisionado

No aprendizado não supervisionado, os dados não possuem, necessariamente, um rótulo indicando qual deveria ser a resposta correta. O algoritmo procura identificar estruturas, padrões ou relações presentes nos próprios dados.

Esse tipo de aprendizado é especialmente útil quando se deseja explorar uma base de dados e descobrir grupos ou padrões que não foram definidos previamente.

Exemplos de aplicações:

- segmentação de clientes com características semelhantes;
- agrupamento de documentos por assunto;
- identificação de padrões de comportamento;
- redução da dimensionalidade de dados;
- exploração de conjuntos de dados sem classes previamente definidas.

### Exemplos de modelos não supervisionados

Alguns algoritmos utilizados em aprendizado não supervisionado são:

- **k-Means**;
- **DBSCAN**;
- **Hierarchical Clustering**;
- **Gaussian Mixture Models (GMM)**;
- **PCA (Principal Component Analysis)**, utilizado principalmente para redução de dimensionalidade.

Um ponto importante é que nem todo método não supervisionado tem como objetivo criar grupos. Técnicas como PCA, por exemplo, procuram representar os dados de uma forma mais compacta, preservando o máximo possível da informação relevante.

## Aprendizado por reforço

No aprendizado por reforço, um agente aprende a tomar decisões interagindo com um ambiente. A cada ação, o agente pode receber uma **recompensa** ou uma **penalização**, e utiliza essas experiências para aprender uma estratégia de decisão.

Diferentemente do aprendizado supervisionado, não é necessário fornecer a resposta correta para cada situação. O agente aprende com as consequências de suas ações.

Exemplos de aplicações:

- jogos eletrônicos;
- controle de robôs;
- sistemas de navegação e tomada de decisão;
- gerenciamento de recursos;
- otimização de estratégias em ambientes simulados.

### Exemplos de métodos de aprendizado por reforço

Entre os métodos conhecidos estão:

- **Q-Learning**;
- **SARSA**;
- **Deep Q-Network (DQN)**;
- métodos baseados em **Policy Gradient**;
- **Actor-Critic**.

Em problemas mais complexos, redes neurais podem ser utilizadas para representar a política de decisão ou estimar os valores associados às ações.

## Regressão

Regressão é uma tarefa de aprendizado de máquina utilizada para prever um **valor numérico**. O modelo aprende uma relação entre as variáveis de entrada e uma variável de saída contínua.

Por exemplo, um modelo pode receber informações como área, localização e número de quartos de um imóvel e estimar seu preço.

Outros exemplos incluem:

- previsão de temperatura;
- previsão de consumo de energia;
- estimativa de vendas;
- previsão de tempo de execução de um processo.

Alguns modelos utilizados para regressão são Regressão Linear, Árvores de Decisão, Random Forest, Support Vector Regression e Redes Neurais.

## Classificação

Classificação é uma tarefa na qual o modelo deve atribuir uma **classe ou categoria** a cada exemplo de entrada.

Por exemplo, um modelo pode analisar características de uma mensagem e classificá-la como **spam** ou **não spam**. Em outro problema, uma imagem pode ser classificada como pertencente à categoria de gato, cachorro ou outra classe.

A classificação pode ser:

- **binária**, quando existem duas classes;
- **multiclasse**, quando existem mais de duas classes possíveis;
- **multirrótulo**, quando um mesmo exemplo pode pertencer simultaneamente a várias classes.

Alguns modelos utilizados para classificação são Regressão Logística, k-NN, Árvores de Decisão, Random Forest, SVM e Redes Neurais.

## Clusterização

Clusterização, ou agrupamento, é uma tarefa de aprendizado não supervisionado cujo objetivo é organizar os dados em grupos de acordo com sua similaridade.

Nesse caso, não são fornecidas classes corretas durante o treinamento. O algoritmo procura identificar uma estrutura de grupos a partir das características dos próprios dados.

Por exemplo, uma empresa pode utilizar clusterização para separar clientes em grupos com comportamentos semelhantes, mesmo sem possuir previamente uma classificação desses clientes.

Alguns algoritmos de clusterização são **k-Means**, **DBSCAN** e métodos de **agrupamento hierárquico**.

## Relação entre os principais tipos de tarefas

| Tarefa | Tipo de aprendizado mais comum | Saída | Exemplo |
|---|---|---|---|
| Regressão | Supervisionado | Valor numérico | Preço de um imóvel |
| Classificação | Supervisionado | Classe/categoria | Spam ou não spam |
| Clusterização | Não supervisionado | Grupos | Segmentação de clientes |

Assim, **regressão** e **classificação** normalmente trabalham com dados rotulados, enquanto **clusterização** normalmente procura descobrir grupos sem que os rótulos sejam previamente fornecidos.
