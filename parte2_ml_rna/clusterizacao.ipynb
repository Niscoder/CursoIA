```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import pandas as pd
from sklearn.cluster import DBSCAN, KMeans, AffinityPropagation
import seaborn as snb
```

```python
df = pd.read_csv('dados_limpos.csv')

encoder = LabelEncoder()
new_df = pd.DataFrame()

for column in df.columns:
    if not column in ["idade", "periodo", "altura"]:
        new_df[column] = encoder.fit_transform(df[column])

    else:
        new_df[column] = df[column]

print(new_df.head())
```

```python
clt = KMeans(3)
clt.fit_predict(new_df)
```

```python
plot_df = new_df.copy()
plot_df["cluster"] = clt.fit_predict(new_df)
```

```python
print(len(df.columns))
```

```python


snb.pairplot(plot_df, hue="cluster", palette="viridis").savefig("pairplot.png" ,dpi=300)
```
