```python
import torch
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, root_mean_squared_error, mean_squared_error
from torch import Tensor
from torch.nn import Sequential, Linear, ReLU, MSELoss
from torch.utils.data import DataLoader
from torch.optim import Adam
```

```python
dataset = load_diabetes()

X_treino, X_teste, y_treino, y_teste = train_test_split(dataset.data, dataset.target, test_size=0.2, random_state=42)
```

```python
model = Sequential(
    Linear(10, 16),
    ReLU(),
    Linear(16,16),
    ReLU(),
    Linear(16, 1)
)
```

```python
optmizer = Adam(model.parameters(), lr=0.001)
criterion = MSELoss()
max_epochs = 100
batch = 8
```

```python
total_batchs = len(X_treino) // batch
last_batch = len(X_treino) % batch
has_last_b = last_batch > 0

print(total_batchs)
print(last_batch)
```

```python
def train_model():
    for i_epochs in range(max_epochs):
        for i_batchs in range(total_batchs):
            batch_x = Tensor(X_treino[(i_batchs * 8):((i_batchs + 1) * 8)])
            batch_y = Tensor(y_treino[(i_batchs * 8):((i_batchs + 1) * 8)])
            optmizer.zero_grad()
            output = model(batch_x)
            loss = criterion(output, batch_y)
            loss.backward()
            optmizer.step()

def eval_model():
    with torch.no_grad():
        X_teste_tensor = Tensor(X_teste)
        output = np.asarray(model(X_teste_tensor).cpu())
        print(f"R^2: {r2_score(y_teste, output)}")
        print(f"RMSE: {root_mean_squared_error(y_teste, output)}")
        print(f"MSE: {mean_squared_error(y_teste, output)}")
```

```python
train_model()
eval_model()
```
