## Variáveis

Uma variável é um nome utilizado para referenciar um objeto.

Em Python, todos os valores são objetos. Cada objeto possui um tipo e uma identidade (id) que o identifica durante sua existência.

## Tipos de dados

O tipo de um objeto determina quais valores ele pode representar e quais operações podem ser realizadas sobre ele.

### String (str)

Representa textos.

```python
nome = "João"
cidade = "Londrina"

print(nome)
print(cidade)
```

### Inteiro (int)

Representa números inteiros.

```python
idade = 25
ano = 2026
temperatura = -5

print(idade)
print(ano)
print(temperatura)
```

### Ponto flutuante (float)

Representa números com casas decimais.

```python
altura = 1.75
preco = 19.90
temperatura = -2.5

print(altura)
print(preco)
print(temperatura)
```

### Booleano (bool)

Representa valores lógicos: True ou False.

```python
maior_de_idade = True
possui_carteira = False

print(maior_de_idade)
print(possui_carteira)
```

## Verificando o tipo

A função type() retorna o tipo do objeto referenciado pela variável.

```python
nome = "João"
idade = 25

print(type(nome))
print(type(idade))
```

### Identidade do objeto

A função id() retorna um identificador associado à identidade do objeto.

```python
nome = "André"

print(id(nome))
```

Duas variáveis podem referenciar o mesmo objeto:

```python
nome = "João"
outro_nome = nome

print(nome)
print(outro_nome)

print(id(nome))
print(id(outro_nome))
```

Nesse caso, nome e outro_nome referenciam o mesmo objeto, portanto possuem o mesmo id.

### Alterando uma variável

Ao atribuir um novo valor, a variável pode passar a referenciar outro objeto.

```python
idade = 25
print(id(idade))

idade = 26
print(id(idade))
```

A variável idade passou a referenciar o objeto correspondente ao valor 26.

## Operações aritméticas

Python possui operadores para realizar operações matemáticas com valores numéricos.

## Operadores aritméticos

| Operador | Operação         |
| -------- | ---------------- |
| `+`      | Adição           |
| `-`      | Subtração        |
| `*`      | Multiplicação    |
| `/`      | Divisão          |
| `//`     | Divisão inteira  |
| `%`      | Resto da divisão |
| `**`     | Exponenciação    |

### Adição (`+`)

Realiza a soma de dois valores.

```python
a = 10
b = 5

resultado = a + b

print(resultado)
```

### Subtração (`-`)

Realiza a subtração de dois valores.

```python
a = 10
b = 5

resultado = a - b

print(resultado)
```

### Multiplicação (`*`)

Multiplica dois valores.

```python
a = 10
b = 5

resultado = a * b

print(resultado)
```

### Divisão (`/`)

Realiza uma divisão e retorna o resultado como `float`.

```python
a = 10
b = 4

resultado = a / b

print(resultado)
print(type(resultado))
```

### Divisão inteira (`//`)

Realiza a divisão descartando a parte fracionária do resultado.

```python
a = 10
b = 4

resultado = a // b

print(resultado)
```

### Resto da divisão (`%`)

Retorna o resto de uma divisão.

```python
a = 10
b = 4

resto = a % b

print(resto)
```

Esse operador é bastante utilizado para verificar, por exemplo, se um número é par:

```python
numero = 10

print(numero % 2 == 0)
```

### Exponenciação (`**`)

Eleva um número a uma determinada potência.

```python
base = 2
expoente = 3

resultado = base ** expoente

print(resultado)
```

## Precedência de operadores

Assim como na matemática, algumas operações possuem prioridade sobre outras.

A multiplicação, divisão, divisão inteira, resto e exponenciação são realizadas antes da adição e subtração.

```python
resultado = 2 + 3 * 4

print(resultado)
```

Parênteses podem ser utilizados para alterar a ordem das operações:

```python
resultado = (2 + 3) * 4

print(resultado)
```
