## Classes

Classes permitem criar estruturas que agrupam dados e comportamentos relacionados.

Uma classe pode ser utilizada como um modelo para criar objetos. Os objetos criados a partir de uma classe são chamados de instâncias.

Por exemplo, podemos criar uma classe para representar uma pessoa:

````python
class Pessoa:
    pass
````

A partir dela, podemos criar objetos:

````python
pessoa1 = Pessoa()
pessoa2 = Pessoa()
````

`pessoa1` e `pessoa2` são objetos distintos, mas ambos são instâncias da classe `Pessoa`.

### Atributos

Atributos são dados associados a um objeto.

Podemos criar atributos diretamente em uma instância:

````python
pessoa = Pessoa()

pessoa.nome = "Carlos"
pessoa.idade = 25
````

Agora podemos acessar esses atributos:

````python
print(pessoa.nome)
print(pessoa.idade)
````

Cada objeto pode possuir seus próprios valores:

````python
pessoa1 = Pessoa()
pessoa1.nome = "Carlos"

pessoa2 = Pessoa()
pessoa2.nome = "Maria"

print(pessoa1.nome)
print(pessoa2.nome)
````

Embora os dois objetos sejam instâncias da mesma classe, cada um possui seus próprios atributos.

### `__init__`

Normalmente queremos que os atributos sejam definidos quando o objeto é criado.

Para isso, utilizamos o método especial `__init__()`:

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
````

O parâmetro `self` representa a própria instância. Por isso:

```python
self.nome = nome
```

significa que o valor recebido pelo parâmetro `nome` será armazenado no atributo `nome` daquela instância.

Agora podemos criar uma pessoa já informando seus dados:

````python
pessoa = Pessoa("Carlos", 25)

print(pessoa.nome)
print(pessoa.idade)
````

### Métodos

Métodos são funções definidas dentro de uma classe e representam comportamentos dos objetos.

````python
class Pessoa:
    def __init__(self, nome):
        self.nome = nome

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome}!")
````

Podemos chamar o método através do objeto:

````python
pessoa = Pessoa("Carlos")

pessoa.apresentar()
````

Assim como nas funções, métodos podem receber parâmetros e retornar valores:

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def fazer_aniversario(self):
        self.idade += 1

    def obter_idade(self):
        return self.idade

pessoa = Pessoa("Carlos", 25)

pessoa.fazer_aniversario()

print(pessoa.obter_idade())
````

### Atributos e métodos de uma instância

Os atributos representam dados do objeto, enquanto os métodos representam comportamentos.

Por exemplo:

````python
class Conta():
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        self.saldo += valor

    def consultar_saldo(self):
        return self.saldo
````

Nesse exemplo:

```text
titular          → atributo
saldo            → atributo
depositar()      → método
consultar_saldo() → método
```

Utilizando a classe:

````python
conta = Conta("Carlos", 1000)

conta.depositar(500)

print(conta.consultar_saldo())
````
