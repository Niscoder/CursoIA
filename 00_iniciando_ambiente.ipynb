## Criando projeto

Crie um diretório para o projeto. Abra com o VSCode e confirme que ele é um diretório confiável.

## Criando ambiente virtual

O `venv` cria um ambiente Python isolado para o projeto, evitando conflitos entre dependências.

### 1. Criar o ambiente virtual

Comando:

```bash
python -m venv <nome do ambiente>
```

Exemplo:

```bash
python -m venv .venv
```

Isso cria uma pasta `.venv` contendo o ambiente virtual do projeto. Por boas práticas, utilizar o prefixo `.` no. nome do ambiente

### 2. Ativar o ambiente virtual

Comando:

```bash
.\<nome do ambiente>\Scripts\activate
```

Após a ativação, o terminal indicará o ambiente ativo, normalmente com `(<nome do ambiente>)` no início da linha. Exemplo `(.venv)`.


## Git

O Git é um sistema de controle de versão utilizado para registrar e compartilhar as alterações realizadas no projeto.

### 1. Clonar o projeto

Clone o repositório para sua máquina utilizando a URL do projeto:

```bash
git clone <url do repositório>
```

Exemplo:

```bash
git clone https://github.com/usuario/projeto.git
```

Isso cria uma pasta contendo os arquivos e o histórico do projeto.

Entre no diretório do projeto:

```bash
cd <nome do projeto>
```

### 2. Adicionar as alterações

Após realizar as alterações no projeto, adicione os arquivos modificados:

```bash
git add .
```

O `.` indica que todas as alterações do diretório atual devem ser adicionadas.

### 3. Criar um commit

Registre as alterações adicionadas:

```bash
git commit -m "<Descrição das alterações>"
```

Exemplo:

```bash
git commit -m "Adiciona configuração inicial do projeto"
```

O commit cria um registro das alterações no histórico do Git.

### 4. Enviar as alterações

Envie os commits para o repositório remoto:

```bash
git push
```

Após o `push`, as alterações estarão disponíveis no repositório remoto.
